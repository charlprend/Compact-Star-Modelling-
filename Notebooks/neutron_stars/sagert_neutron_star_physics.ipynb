# 3. Pure Neutron Stars

## A. Nonrelativistic Polytropic Case

Sagert begins by considering an idealised neutron star composed entirely of neutrons, described as a degenerate Fermi gas at zero temperature.

This extends the approach used for white dwarfs, where electrons provided the degeneracy pressure while nuclei contributed most of the mass.

In the pure neutron star model, neutrons provide both the degeneracy pressure and the mass of the stellar matter.

Initially, the neutrons are treated as noninteracting particles. Contributions from protons, electrons and interactions between nucleons are introduced in later sections.

Although this section considers the nonrelativistic polytropic case, Sagert first establishes the full relativistic expressions for neutron energy density and pressure before taking their nonrelativistic limits.

### Equation (66): Neutron Number Density

**How is the number density of neutrons related to the neutron Fermi momentum?**

Neutrons are spin-$1/2$ fermions and obey the Pauli exclusion principle.

At zero temperature, the available momentum states are occupied up to the neutron Fermi momentum, $k_n$.

Sagert drops the subscript $F$ from the Fermi momentum notation, so $k_n$ represents the neutron Fermi momentum throughout this section.

From the phase space treatment introduced in Equations (10) to (13), the neutron number density is

$$
n_n=
\frac{g}{(2\pi\hbar)^3}
\int_0^{k_n}4\pi k^2\,dk
$$

where $g=2$ accounts for the two neutron spin states.

Substituting $g=2$ and moving the constants outside the integral:

$$
n_n=
\left(\frac{2}{(2\pi\hbar)^3}\right)
4\pi\int_0^{k_n}k^2\,dk
$$

Combining the numerical factors:

$$
n_n=
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}k^2\,dk
$$

Integrating:

$$
\int k^2\,dk=\frac{k^3}{3}
$$

Therefore:

$$
n_n=
\frac{8\pi}{(2\pi\hbar)^3}
\left[\frac{k^3}{3}\right]_0^{k_n}
$$

Evaluating the limits:

$$
n_n=
\frac{8\pi}{(2\pi\hbar)^3}
\left(\frac{k_n^3}{3}\right)
$$

Expand the denominator:

$$
(2\pi\hbar)^3=8\pi^3\hbar^3
$$

Giving:

$$
n_n=
\frac{8\pi}{8\pi^3\hbar^3}
\left(\frac{k_n^3}{3}\right)
$$

Cancel the common factor $8\pi$:

$$
n_n=
\frac{1}{\pi^2\hbar^3}
\left(\frac{k_n^3}{3}\right)
$$

Therefore:

$$
{
n_n=\frac{k_n^3}{3\pi^2\hbar^3}
}
$$

This is Sagert's Equation (66).

The neutron number density is proportional to the cube of the Fermi momentum:

$$
n_n\propto k_n^3
$$

Increasing the number density forces neutrons into progressively higher momentum states because the lower momentum states are already occupied.

### Equation (67): Neutron Mass Density

**How is neutron number density related to mass density?**

For this idealised model, the stellar matter consists entirely of neutrons.

The mass density is therefore the number of neutrons per unit volume multiplied by the mass of an individual neutron:

$$
\boxed{
\rho=n_nm_n
}
$$

This is Equation (67), where $m_n$ is the neutron rest mass.

Substituting Equation (66):

$$
\rho=
\frac{k_n^3}{3\pi^2\hbar^3}m_n
$$

Therefore:

$$
{
\rho=\frac{m_nk_n^3}{3\pi^2\hbar^3}
}
$$

The associated rest mass energy density is

$$
\rho c^2=
\frac{m_nc^2k_n^3}{3\pi^2\hbar^3}
$$

This quantity represents the rest mass energy per unit volume.

The full neutron energy density, derived next, also includes the kinetic energy associated with the occupied momentum states.

### Equation (68): Neutron Energy Density

**How do we calculate the total energy density of a degenerate neutron gas?**

As with the electron gas in the white dwarf model, the total energy density is obtained by integrating the energy of the occupied momentum states up to the Fermi momentum.

For a neutron with momentum $k$, the relativistic energy is

$$
E_n(k)=\sqrt{k^2c^2+m_n^2c^4}
$$

This includes both the neutron rest mass energy and its kinetic energy.

For a degenerate neutron gas at zero temperature:

$$
\epsilon_n=
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}E_n(k)k^2\,dk
$$

Substituting the relativistic energy expression:

$$
\epsilon_n=
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}
k^2\sqrt{k^2c^2+m_n^2c^4}\,dk
$$

Expand the denominator:

$$
(2\pi\hbar)^3=8\pi^3\hbar^3
$$

Therefore:

$$
\frac{8\pi}{(2\pi\hbar)^3}
=
\frac{8\pi}{8\pi^3\hbar^3}
=
\frac{1}{\pi^2\hbar^3}
$$

Giving:

$$
\epsilon_n=
\frac{1}{\pi^2\hbar^3}
\int_0^{k_n}
(k^2c^2+m_n^2c^4)^{1/2}k^2\,dk
$$

#### Introducing a Dimensionless Momentum

As in the white dwarf derivation, introduce the dimensionless variable

$$
u=\frac{k}{m_nc}
$$

Rearranging:

$$
k=m_ncu
$$

Differentiating:

$$
dk=m_nc\,du
$$

The squared momentum becomes

$$
k^2=m_n^2c^2u^2
$$

Inside the square root:

$$
k^2c^2+m_n^2c^4
=
m_n^2c^4u^2+m_n^2c^4
$$

Factor out $m_n^2c^4$:

$$
=m_n^2c^4(1+u^2)
$$

Taking the square root:

$$
\sqrt{k^2c^2+m_n^2c^4}
=
m_nc^2\sqrt{1+u^2}
$$

Substituting these expressions into the energy density integral:

$$
\epsilon_n=
\frac{1}{\pi^2\hbar^3}
\int_0^{k_n/(m_nc)}
(m_n^2c^2u^2)
(m_nc^2\sqrt{1+u^2})
(m_nc\,du)
$$

Collect the factors of neutron mass and the speed of light:

$$
m_n^2m_nm_n=m_n^4
$$

$$
c^2c^2c=c^5
$$

Therefore:

$$
\epsilon_n=
\frac{m_n^4c^5}{\pi^2\hbar^3}
\int_0^{k_n/(m_nc)}
u^2\sqrt{1+u^2}\,du
$$

Define the characteristic neutron energy density scale:

$$
\epsilon_0=
\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

And the dimensionless neutron Fermi momentum:

$$
x=\frac{k_n}{m_nc}
$$

These are the quantities formally defined in Equation (70).

The energy density becomes

$$
{
\epsilon_n(x)=
\epsilon_0
\int_0^x u^2\sqrt{1+u^2}\,du
}
$$

#### Evaluating the Energy Density Integral

We need to evaluate

$$
I_\epsilon=
\int u^2\sqrt{1+u^2}\,du
$$

Because the integrand contains $\sqrt{1+u^2}$, use a hyperbolic substitution:

$$
u=\sinh\theta
$$

Differentiating:

$$
du=\cosh\theta\,d\theta
$$

Using the hyperbolic identity

$$
1+\sinh^2\theta=\cosh^2\theta
$$

we obtain

$$
\sqrt{1+u^2}
=
\sqrt{1+\sinh^2\theta}
=
\cosh\theta
$$

Since $u\geq0$, we take $\theta\geq0$, so $\cosh\theta$ is positive.

Substitute into the integral:

$$
u^2\sqrt{1+u^2}\,du
=
\sinh^2\theta\cosh\theta\cosh\theta\,d\theta
$$

Therefore:

$$
I_\epsilon=
\int\sinh^2\theta\cosh^2\theta\,d\theta
$$

Write the product as a square:

$$
I_\epsilon=
\int(\sinh\theta\cosh\theta)^2\,d\theta
$$

Use the double angle identity:

$$
\sinh(2\theta)=2\sinh\theta\cosh\theta
$$

Divide by 2:

$$
\sinh\theta\cosh\theta=
\frac12\sinh(2\theta)
$$

Square both sides:

$$
(\sinh\theta\cosh\theta)^2
=
\left(\frac12\sinh(2\theta)\right)^2
$$

Therefore:

$$
=
\frac14\sinh^2(2\theta)
$$

The integral becomes

$$
I_\epsilon=
\frac14\int\sinh^2(2\theta)\,d\theta
$$

Now use the identity

$$
\cosh(4\theta)=1+2\sinh^2(2\theta)
$$

Subtract 1:

$$
\cosh(4\theta)-1=2\sinh^2(2\theta)
$$

Divide by 2:

$$
\sinh^2(2\theta)=
\frac{\cosh(4\theta)-1}{2}
$$

Substitute:

$$
I_\epsilon=
\frac14
\int\frac{\cosh(4\theta)-1}{2}\,d\theta
$$

Combining the constants:

$$
I_\epsilon=
\frac18
\int[\cosh(4\theta)-1]\,d\theta
$$

#### Integrating with the Reverse Chain Rule

For the first term:

$$
\frac{d}{d\theta}\sinh(4\theta)
=
4\cosh(4\theta)
$$

Therefore:

$$
\int\cosh(4\theta)\,d\theta
=
\frac14\sinh(4\theta)
$$

For the second term:

$$
\int(-1)\,d\theta=-\theta
$$

Combining the results:

$$
I_\epsilon=
\frac18\left[
\frac14\sinh(4\theta)-\theta
\right]+C
$$

#### Returning to the Original Variable

We have

$$
u=\sinh\theta
$$

Therefore:

$$
\theta=\sinh^{-1}(u)
$$

We must also express $\sinh(4\theta)$ in terms of $u$.

Using the double angle identity:

$$
\sinh(4\theta)=
2\sinh(2\theta)\cosh(2\theta)
$$

And

$$
\sinh(2\theta)=
2\sinh\theta\cosh\theta
$$

Substitute:

$$
\sinh(4\theta)=
4\sinh\theta\cosh\theta\cosh(2\theta)
$$

Divide by 4:

$$
\frac14\sinh(4\theta)=
\sinh\theta\cosh\theta\cosh(2\theta)
$$

From our substitution:

$$
\sinh\theta=u
$$

$$
\cosh\theta=\sqrt{1+u^2}
$$

The remaining factor is found using

$$
\cosh(2\theta)=1+2\sinh^2\theta
$$

Therefore:

$$
\cosh(2\theta)=1+2u^2
$$

Substitute all three expressions:

$$
\frac14\sinh(4\theta)=
u\sqrt{1+u^2}(1+2u^2)
$$

Expand the bracket:

$$
u(1+2u^2)=u+2u^3
$$

Giving:

$$
\frac14\sinh(4\theta)=
(2u^3+u)\sqrt{1+u^2}
$$

Returning to the integral:

$$
I_\epsilon=
\frac18
\left[
(2u^3+u)\sqrt{1+u^2}
-\sinh^{-1}(u)
\right]+C
$$

#### Applying the Integration Limits

The neutron energy density requires the definite integral from $u=0$ to $u=x$:

$$
\int_0^x u^2\sqrt{1+u^2}\,du
=
\frac18
\left[
(2u^3+u)\sqrt{1+u^2}
-\sinh^{-1}(u)
\right]_0^x
$$

At the lower limit:

$$
u=0
$$

$$
(2(0)^3+0)\sqrt{1+0^2}=0
$$

and

$$
\sinh^{-1}(0)=0
$$

The lower limit therefore contributes zero.

At the upper limit, $u=x$:

$$
\int_0^x u^2\sqrt{1+u^2}\,du
=
\frac18
\left[
(2x^3+x)\sqrt{1+x^2}
-\sinh^{-1}(x)
\right]
$$

Recall that

$$
\epsilon_n(x)=
\epsilon_0\int_0^x u^2\sqrt{1+u^2}\,du
$$

Therefore:

$$
{
\epsilon_n(x)=
\frac{\epsilon_0}{8}
\left[
(2x^3+x)\sqrt{1+x^2}
-\sinh^{-1}(x)
\right]
}
$$

This reproduces Sagert's Equation (68).

The expression gives the full energy density of the ideal degenerate neutron gas, including its rest mass and kinetic energy.

Unlike the white dwarf model, where the nuclear rest mass energy and electron energy density were calculated separately, the neutron energy integral already includes the rest mass energy of the neutrons.


### Equation (69): Neutron Degeneracy Pressure

**How do we calculate the pressure produced by the degenerate neutron gas?**

As with the electron gas, the pressure is obtained by integrating the momentum and velocity contributions from the occupied particle states.

For an isotropic momentum distribution:

$$
p=
\frac13
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}kv(k)k^2\,dk
$$

The factor $1/3$ arises because the momentum distribution is isotropic. There is no preferred spatial direction, so the pressure contributions are equivalent along the three spatial axes.

For a relativistic neutron, the velocity is

$$
v(k)=\frac{kc^2}{E_n(k)}
$$

where

$$
E_n(k)=\sqrt{k^2c^2+m_n^2c^4}
$$

Substitute the velocity into the pressure expression:

$$
p(k_n)=
\frac13
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}
k\left(\frac{kc^2}{E_n(k)}\right)k^2\,dk
$$

Combine the powers of momentum:

$$
k\cdot k\cdot k^2=k^4
$$

Therefore:

$$
p(k_n)=
\frac13
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}
\frac{k^4c^2}{E_n(k)}\,dk
$$

Substituting the relativistic energy:

$$
p(k_n)=
\frac13
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}
\frac{k^4c^2}
{\sqrt{k^2c^2+m_n^2c^4}}\,dk
$$

Equivalently, write the denominator as a negative exponent:

$$
p(k_n)=
\frac13
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_n}
(k^2c^2+m_n^2c^4)^{-1/2}
k^4c^2\,dk
$$

Simplifying the constant factor:

$$
\frac13\frac{8\pi}{(2\pi\hbar)^3}
=
\frac{1}{3\pi^2\hbar^3}
$$

Therefore:

$$
p(k_n)=
\frac{1}{3\pi^2\hbar^3}
\int_0^{k_n}
(k^2c^2+m_n^2c^4)^{-1/2}
k^4c^2\,dk
$$

#### Introducing the Dimensionless Momentum

As before, define

$$
u=\frac{k}{m_nc}
$$

Therefore:

$$
k=m_ncu
$$

$$
dk=m_nc\,du
$$

and

$$
k^4=m_n^4c^4u^4
$$

Multiplying by $c^2$:

$$
k^4c^2=m_n^4c^6u^4
$$

The energy denominator becomes

$$
\sqrt{k^2c^2+m_n^2c^4}
=
m_nc^2\sqrt{1+u^2}
$$

Substituting into the pressure integral:

$$
p=
\frac{1}{3\pi^2\hbar^3}
\int_0^x
\frac{m_n^4c^6u^4}
{m_nc^2\sqrt{1+u^2}}
m_nc\,du
$$

Simplify the numerator and denominator:

$$
\frac{m_n^4c^6}{m_nc^2}
=
m_n^3c^4
$$

Multiplying by $m_nc$ from $dk$:

$$
m_n^3c^4\cdot m_nc=m_n^4c^5
$$

Therefore:

$$
p=
\frac{m_n^4c^5}{3\pi^2\hbar^3}
\int_0^x\frac{u^4}{\sqrt{1+u^2}}\,du
$$

Using

$$
\epsilon_0=
\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

the pressure becomes

$$
{
p(x)=
\frac{\epsilon_0}{3}
\int_0^x u^4(1+u^2)^{-1/2}\,du
}
$$

We need to evaluate

I_p=

Use the same hyperbolic substitution:

The factors of $\cosh\theta$ cancel:

Subtract 1:

### Equation (70): Dimensionless Momentum and Energy Density Scale

Sagert defines

and

$$
{
\epsilon_0=
\frac{m_n^4c^5}{\pi^2\hbar^3}
}
$$

The dimensionless momentum $x$ measures the neutron Fermi momentum relative to the neutron rest mass momentum $m_nc$.

For

$$
x\ll1
$$

the neutrons are nonrelativistic.

For

$$
x\gg1
$$

the neutrons are ultrarelativistic.

The quantity $\epsilon_0$ provides the characteristic energy density scale for the neutron gas. It has units of energy per unit volume, $\mathrm{J\,m^{-3}}$.

The same scale appears in both the energy density and pressure expressions.

### Physical Interpretation

Equations (66) to (70) describe an ideal degenerate neutron gas.

The mathematical procedure is analogous to the degenerate electron gas developed in the white dwarf section.

For the white dwarf, electrons provide most of the pressure while nuclei provide most of the mass.

For the pure neutron gas, neutrons provide both the pressure and the energy density.

The central expressions are

**Neutron number density:**

$$
n_n=\frac{k_n^3}{3\pi^2\hbar^3}
$$

**Neutron energy density:**

$$
\epsilon(x)=
\frac{\epsilon_0}{8}
\left[
(2x^3+x)\sqrt{1+x^2}
-\sinh^{-1}(x)
\right]
$$

**Neutron degeneracy pressure:**

$$
p(x)=
\frac{\epsilon_0}{24}
\left[
(2x^3-3x)\sqrt{1+x^2}
+3\sinh^{-1}(x)
\right]
$$

Both the energy density and pressure are functions of the dimensionless neutron Fermi momentum $x$.

The energy density integral includes the neutron rest mass energy, so an additional $\rho c^2$ contribution must not be added to this pure neutron gas energy density.

The model also neglects interactions between neutrons and the presence of other particles. These approximations will need to be reconsidered when developing more realistic neutron star equations of state.




### Equation (71): Nonrelativistic Energy Density and Pressure

From Equation (70), the dimensionless neutron Fermi momentum is

$$
x=\frac{k_n}{m_nc}
$$

For nonrelativistic neutrons,

$$
k_n\ll m_nc
$$

Therefore,

$$
\boxed{x\ll1}
$$

Since the integration variable runs from $u=0$ to $u=x$, we also have $u\ll1$ throughout the integrals.

The full expressions derived previously were

$$
\epsilon(x)=\epsilon_0\int_0^x u^2\sqrt{1+u^2}\,du
$$

and

$$
p(x)=\frac{\epsilon_0}{3}
\int_0^x\frac{u^4}{\sqrt{1+u^2}}\,du
$$

where

$$
\epsilon_0=\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

We will use these integral forms because the nonrelativistic approximation can be applied directly to their integrands.

#### Part A: Energy Density

Starting with

$$
\epsilon(x)=\epsilon_0\int_0^x u^2\sqrt{1+u^2}\,du
$$

Since $u\ll1$, we know that $u^2$ is much smaller than 1.

Using the leading term of the binomial expansion,

$$
(1+u^2)^{1/2}
=
1+\frac12u^2-\frac18u^4+\cdots
$$

In the nonrelativistic limit,

$$
\sqrt{1+u^2}\simeq1
$$

Substituting this approximation:

$$
\epsilon(x)\simeq\epsilon_0\int_0^x u^2(1)\,du
$$

Therefore,

$$
\epsilon(x)\simeq\epsilon_0\int_0^x u^2\,du
$$

Using the power rule for integration,

$$
\int u^n\,du=\frac{u^{n+1}}{n+1}+C
$$

we obtain

$$
\int u^2\,du=\frac{u^3}{3}
$$

Therefore,

$$
\epsilon(x)\simeq
\epsilon_0\left[\frac{u^3}{3}\right]_0^x
$$

Evaluate the upper and lower limits:

$$
\epsilon(x)\simeq
\epsilon_0\left(\frac{x^3}{3}-\frac{0^3}{3}\right)
$$

Hence,

$$
\boxed{
\epsilon(x)\simeq\frac{\epsilon_0}{3}x^3
}
$$

#### Why does Sagert also write $\epsilon\simeq\rho c^2$?

From Equation (66),

$$
n_n=\frac{k_n^3}{3\pi^2\hbar^3}
$$

From Equation (67),

$$
\rho=n_nm_n
$$

Therefore,

$$
\rho=\frac{m_nk_n^3}{3\pi^2\hbar^3}
$$

Multiplying by $c^2$ gives the rest mass energy density:

$$
\rho c^2=
\frac{m_nc^2k_n^3}{3\pi^2\hbar^3}
$$

From Equation (70),

$$
x=\frac{k_n}{m_nc}
$$

Rearranging,

$$
k_n=m_ncx
$$

Cubing both sides:

$$
k_n^3=(m_ncx)^3
$$

Therefore,

$$
k_n^3=m_n^3c^3x^3
$$

Substitute into the rest mass energy density:

$$
\rho c^2=
\frac{m_nc^2(m_n^3c^3x^3)}
{3\pi^2\hbar^3}
$$

Collect the powers:

$$
\rho c^2=
\frac{m_n^4c^5x^3}
{3\pi^2\hbar^3}
$$

From Equation (70),

$$
\epsilon_0=\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

Therefore,

$$
\boxed{
\rho c^2=\frac{\epsilon_0}{3}x^3
}
$$

Comparing this with our nonrelativistic energy density,

$$
\boxed{
\epsilon(x)\simeq\rho c^2
=\frac{\epsilon_0}{3}x^3
}
$$

This is the energy density expression in Sagert's Equation (71).

**Physical interpretation**

In the nonrelativistic regime, the neutron kinetic energy is much smaller than the neutron rest mass energy.

The leading contribution to the total energy density is therefore the rest mass energy:

$$
\epsilon\simeq n_nm_nc^2
$$

This does not mean the kinetic energy vanishes. Its contribution is simply smaller than the leading rest mass term.

#### Part B: Degeneracy Pressure

Starting from the pressure integral derived for Equation (69),

$$
p(x)=\frac{\epsilon_0}{3}
\int_0^x\frac{u^4}{\sqrt{1+u^2}}\,du
$$

For $u\ll1$,

$$
\sqrt{1+u^2}\simeq1
$$

Therefore,

$$
\frac{1}{\sqrt{1+u^2}}\simeq1
$$

Substituting the approximation:

$$
p(x)\simeq
\frac{\epsilon_0}{3}\int_0^x u^4(1)\,du
$$

Hence,

$$
p(x)\simeq
\frac{\epsilon_0}{3}\int_0^x u^4\,du
$$

Using the power rule,

$$
\int u^4\,du=\frac{u^5}{5}
$$

Therefore,

$$
p(x)\simeq
\frac{\epsilon_0}{3}
\left[\frac{u^5}{5}\right]_0^x
$$

Evaluate the limits:

$$
p(x)\simeq
\frac{\epsilon_0}{3}
\left(\frac{x^5}{5}-\frac{0^5}{5}\right)
$$

So,

$$
p(x)\simeq\frac{\epsilon_0}{3}\frac{x^5}{5}
$$

Multiplying the denominators,

$$
3\times5=15
$$

Therefore,

$$
\boxed{
p(x)\simeq\frac{\epsilon_0}{15}x^5
}
$$

This is the pressure expression in Sagert's Equation (71).

#### Equation (71): Final Results

For the nonrelativistic pure neutron gas,

$$
\boxed{
\epsilon(x)\simeq\rho c^2
=\frac{\epsilon_0}{3}x^3
}
$$

and

$$
\boxed{
p(x)\simeq\frac{\epsilon_0}{15}x^5
}
$$

The important difference is the dependence on $x$:

$$
\epsilon\propto x^3
$$

$$
p\propto x^5
$$

The energy density is dominated by rest mass energy, whereas the pressure depends on the motion of the neutrons occupying momentum states up to the Fermi momentum.

These different powers will determine the polytropic exponent.

### Equation (72): Deriving the Nonrelativistic Polytropic Equation of State

#### Aim

To eliminate the dimensionless Fermi momentum $x$ and express the neutron degeneracy pressure directly in terms of energy density.

An equation of state relates pressure to the properties of the matter.

For the stellar structure equations, we require a relationship of the form

$$
p=p(\epsilon)
$$

Sagert expresses the nonrelativistic neutron equation of state as a polytrope:

$$
p(\epsilon)=K_{\mathrm{nrel}}\epsilon^\gamma
$$

We need to determine the exponent $\gamma$ and coefficient $K_{\mathrm{nrel}}$.

#### Step 1: Rearranging the Energy Density

From Equation (71),

$$
\epsilon=\frac{\epsilon_0}{3}x^3
$$

We want to make $x^3$ the subject.

Multiply both sides by

$$
\frac{3}{\epsilon_0}
$$

Giving

$$
\epsilon\frac{3}{\epsilon_0}
=
\frac{\epsilon_0}{3}x^3
\frac{3}{\epsilon_0}
$$

On the right hand side, the factors cancel:

$$
\frac{\epsilon_0}{3}\frac{3}{\epsilon_0}=1
$$

Therefore,

$$
{
x^3=\frac{3\epsilon}{\epsilon_0}
}
$$

#### Step 2: Rewriting the Pressure

From Equation (71),

$$
p=\frac{\epsilon_0}{15}x^5
$$

We have an expression for $x^3$, but the pressure depends on $x^5$.

Using the exponent rule,

$$
(a^m)^n=a^{mn}
$$

we can write

$$
x^5=(x^3)^{5/3}
$$

To verify,

$$
(x^3)^{5/3}=x^{3(5/3)}
$$

Since

$$
3\times\frac53=5
$$

we obtain

$$
(x^3)^{5/3}=x^5
$$

Therefore,

$$
p=\frac{\epsilon_0}{15}(x^3)^{5/3}
$$

#### Step 3: Substituting for $x^3$

We already found that

$$
x^3=\frac{3\epsilon}{\epsilon_0}
$$

Substitute into the pressure equation:

$$
{
p=\frac{\epsilon_0}{15}
\left(\frac{3\epsilon}{\epsilon_0}\right)^{5/3}
}
$$

We now have pressure written entirely in terms of energy density and constants.

#### Step 4: Expanding the Fractional Exponent

Use the exponent rule

$$
\left(\frac{a}{b}\right)^n=\frac{a^n}{b^n}
$$

Therefore,

$$
p=\frac{\epsilon_0}{15}
\frac{(3\epsilon)^{5/3}}
{\epsilon_0^{5/3}}
$$

For a product,

$$
(ab)^n=a^nb^n
$$

Therefore,

$$
(3\epsilon)^{5/3}
=
3^{5/3}\epsilon^{5/3}
$$

Substituting:

$$
p=\frac{\epsilon_0}{15}
\frac{3^{5/3}\epsilon^{5/3}}
{\epsilon_0^{5/3}}
$$

Rearrange to group the factors:

$$
p=
\frac{3^{5/3}}{15}
\frac{\epsilon_0}{\epsilon_0^{5/3}}
\epsilon^{5/3}
$$

#### Step 5: Simplifying the Powers of $\epsilon_0$

We have

$$
\frac{\epsilon_0}{\epsilon_0^{5/3}}
$$

The numerator has an exponent of 1:

$$
\frac{\epsilon_0^1}{\epsilon_0^{5/3}}
$$

Using the quotient rule,

$$
\frac{a^m}{a^n}=a^{m-n}
$$

we obtain

$$
\frac{\epsilon_0^1}{\epsilon_0^{5/3}}
=
\epsilon_0^{1-5/3}
$$

Write 1 with denominator 3:

$$
1=\frac33
$$

Therefore,

$$
1-\frac53=\frac33-\frac53
$$

$$
=\frac{3-5}{3}
$$

$$
=-\frac23
$$

So,

$$
\frac{\epsilon_0}{\epsilon_0^{5/3}}
=
\epsilon_0^{-2/3}
$$

Using the negative exponent rule,

$$
a^{-n}=\frac{1}{a^n}
$$

we have

$$
\epsilon_0^{-2/3}
=
\frac{1}{\epsilon_0^{2/3}}
$$

Returning to the pressure expression:

$$
p=
\frac{3^{5/3}}{15}
\frac{1}{\epsilon_0^{2/3}}
\epsilon^{5/3}
$$

Therefore,

$$
{
p=
\frac{3^{5/3}}{15\epsilon_0^{2/3}}
\epsilon^{5/3}
}
$$

We retain the factors $3$ and $15$ in preparation for obtaining Sagert's Equation (73).

#### Step 6: Identifying the Polytropic Constants

The general polytropic form is

$$
p=K\epsilon^\gamma
$$

Comparing with our expression,

$$
p=
\frac{3^{5/3}}{15\epsilon_0^{2/3}}
\epsilon^{5/3}
$$

we identify the polytropic exponent as

$$
{
\gamma=\frac53
}
$$

and the polytropic coefficient as

$$
{
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15\epsilon_0^{2/3}}
}
$$

Therefore,

$$
{
p(\epsilon)=K_{\mathrm{nrel}}\epsilon^{5/3}
}
$$

This reproduces Sagert's Equation (72).

**Physical interpretation**

The exponent $5/3$ follows from the different powers of Fermi momentum in Equation (71):

$$
\epsilon\propto x^3
$$

and

$$
p\propto x^5
$$

Eliminating $x$ gives

$$
p\propto\epsilon^{5/3}
$$

This describes the pressure response of an ideal, nonrelativistic degenerate neutron gas.

As the energy density increases, the pressure increases more rapidly because additional neutrons must occupy progressively higher momentum states.

The coefficient $K_{\mathrm{nrel}}$ contains the physical constants that set the magnitude of the pressure at a given energy density.


### Equation (73): Deriving the Polytropic Constant in Sagert's Form


To express $K_{\mathrm{nrel}}$ in terms of the fundamental physical constants and rearrange the result into the exact form used by Sagert.

We obtained

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15\epsilon_0^{2/3}}
$$

From Equation (70),

$$
\epsilon_0=
\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

We substitute this expression for $\epsilon_0$.

#### Step 1: Substituting the Energy Density Constant

Starting with

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15\epsilon_0^{2/3}}
$$

Substitute

$$
\epsilon_0=
\frac{m_n^4c^5}{\pi^2\hbar^3}
$$

Giving

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}
{15\left(\dfrac{m_n^4c^5}{\pi^2\hbar^3}\right)^{2/3}}
$$

Moving the fraction raised to $2/3$ into the numerator by taking its reciprocal:

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15}
\left(
\frac{\pi^2\hbar^3}{m_n^4c^5}
\right)^{2/3}
$$

#### Step 2: Applying the Fractional Exponent

Using

$$
\left(\frac{ab}{cd}\right)^n
=
\frac{a^nb^n}{c^nd^n}
$$

we obtain

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15}
\frac{(\pi^2)^{2/3}(\hbar^3)^{2/3}}
{(m_n^4)^{2/3}(c^5)^{2/3}}
$$

Apply the exponent rule

$$
(a^m)^n=a^{mn}
$$

to each quantity separately.

**Reduced Planck constant**

$$
(\hbar^3)^{2/3}
=
\hbar^{3(2/3)}
$$

$$
=\hbar^{6/3}
$$

Therefore,

$$
\boxed{(\hbar^3)^{2/3}=\hbar^2}
$$

**Pi**

$$
(\pi^2)^{2/3}
=
\pi^{2(2/3)}
$$

Therefore,

$$
{(\pi^2)^{2/3}=\pi^{4/3}}
$$

**Neutron mass**

$$
(m_n^4)^{2/3}
=
m_n^{4(2/3)}
$$

Therefore,

$$
{(m_n^4)^{2/3}=m_n^{8/3}}
$$

**Speed of light**

$$
(c^5)^{2/3}
=
c^{5(2/3)}
$$

Therefore,

$$
{(c^5)^{2/3}=c^{10/3}}
$$

Substituting these four results:

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}}{15}
\frac{\pi^{4/3}\hbar^2}
{m_n^{8/3}c^{10/3}}
$$

Combining the factors:

$$
{
K_{\mathrm{nrel}}=
\frac{3^{5/3}\pi^{4/3}\hbar^2}
{15m_n^{8/3}c^{10/3}}
}
$$

This is the fully expanded form of the coefficient.

The next step is to rearrange it into Sagert's Equation (73).

#### Step 3: Rearranging the Neutron Mass

Sagert has one factor of $m_n$ outside the brackets, with another factor of $m_n$ inside the brackets raised to $5/3$.

We have

$$
m_n^{8/3}
$$

Using

$$
a^{m+n}=a^ma^n
$$

write

$$
\frac83=1+\frac53
$$

Therefore,

$$
m_n^{8/3}
=
m_n^{1+5/3}
$$

So

$$
{
m_n^{8/3}=m_n(m_n)^{5/3}
}
$$

#### Step 4: Rearranging the Speed of Light

We have

$$
c^{10/3}
$$

We want to write the speed of light as $c^2$ raised to $5/3$.

Using

$$
(a^m)^n=a^{mn}
$$

we obtain

$$
(c^2)^{5/3}=c^{2(5/3)}
$$

Therefore,

$$
{
c^{10/3}=(c^2)^{5/3}
}
$$

Combining the neutron mass and speed of light:

$$
m_n^{5/3}c^{10/3}
=
m_n^{5/3}(c^2)^{5/3}
$$

Both factors have exponent $5/3$, so

$$
{
m_n^{5/3}c^{10/3}
=
(m_nc^2)^{5/3}
}
$$

Therefore,

$$
m_n^{8/3}c^{10/3}
=
m_n(m_nc^2)^{5/3}
$$

#### Step 5: Rearranging the Powers of Pi

Our expanded coefficient contains

$$
\pi^{4/3}
$$

Sagert has $\pi^2$ outside the brackets in the denominator and $\pi^2$ inside the brackets raised to $5/3$.

To obtain his arrangement, write

$$
\pi^{4/3}
=
\frac{\pi^{10/3}}{\pi^2}
$$

We can verify this using the quotient rule:

$$
\frac{\pi^{10/3}}{\pi^2}
=
\pi^{10/3-2}
$$

Write 2 with denominator 3:

$$
2=\frac63
$$

Therefore,

$$
\pi^{10/3-2}
=
\pi^{10/3-6/3}
$$

$$
=\pi^{4/3}
$$

Now use the power rule:

$$
(\pi^2)^{5/3}
=
\pi^{2(5/3)}
$$

$$
=\pi^{10/3}
$$

Therefore,

$$
{
\pi^{4/3}
=
\frac{(\pi^2)^{5/3}}{\pi^2}
}
$$

This is the arrangement needed to reproduce Sagert's expression.

#### Step 6: Combining the Rearranged Factors

Return to our expanded coefficient:

$$
K_{\mathrm{nrel}}=
\frac{3^{5/3}\pi^{4/3}\hbar^2}
{15m_n^{8/3}c^{10/3}}
$$

Substitute

$$
\pi^{4/3}=
\frac{(\pi^2)^{5/3}}{\pi^2}
$$

to obtain

$$
K_{\mathrm{nrel}}=
\frac{
3^{5/3}
\left(\dfrac{(\pi^2)^{5/3}}{\pi^2}\right)
\hbar^2
}
{15m_n^{8/3}c^{10/3}}
$$

Move the factor $\pi^2$ into the denominator:

$$
K_{\mathrm{nrel}}=
\frac{
3^{5/3}(\pi^2)^{5/3}\hbar^2
}
{15\pi^2m_n^{8/3}c^{10/3}}
$$

From the previous steps,

$$
m_n^{8/3}c^{10/3}
=
m_n(m_nc^2)^{5/3}
$$

Substitute:

$$
K_{\mathrm{nrel}}=
\frac{
3^{5/3}(\pi^2)^{5/3}\hbar^2
}
{15\pi^2m_n(m_nc^2)^{5/3}}
$$

Now group the factors which have the same exponent:

$$
3^{5/3}(\pi^2)^{5/3}
=
(3\pi^2)^{5/3}
$$

Therefore,

$$
K_{\mathrm{nrel}}=
\frac{
\hbar^2(3\pi^2)^{5/3}
}
{15\pi^2m_n(m_nc^2)^{5/3}}
$$

Separate the factor outside the brackets:

$$
K_{\mathrm{nrel}}=
\frac{\hbar^2}{15\pi^2m_n}
\frac{(3\pi^2)^{5/3}}
{(m_nc^2)^{5/3}}
$$

Use

$$
\frac{a^n}{b^n}
=
\left(\frac{a}{b}\right)^n
$$

Therefore,

$$
{
K_{\mathrm{nrel}}=
\frac{\hbar^2}{15\pi^2m_n}
\left(
\frac{3\pi^2}{m_nc^2}
\right)^{5/3}
}
$$

Sagert gives

#### Physical Meaning

The neutron Fermi momentum determines how many momentum states are occupied.

In the nonrelativistic regime, the energy density is dominated by neutron rest mass energy and scales as

$$
\epsilon\propto x^3
$$

The neutron degeneracy pressure scales as

$$
p\propto x^5
$$

Eliminating the Fermi momentum produces

$$
p\propto\epsilon^{5/3}
$$

The exponent $\gamma=5/3$ is characteristic of a nonrelativistic ideal degenerate Fermi gas.

The coefficient $K_{\mathrm{nrel}}$ sets the magnitude of the pressure and depends on the neutron mass, the reduced Planck constant and the speed of light.

This model assumes a zero temperature, noninteracting neutron gas. Its nonrelativistic equation of state is valid only where the neutron Fermi momentum is much smaller than $m_nc$.


### Figure 8: Calculated Masses and Radii for Neutron Stars Using the Nonrelativistic Polytrope

How do the mass and radius of a neutron star change with increasing central pressure, and how do general relativistic corrections alter the Newtonian predictions?

Figure 8 compares two treatments of stellar hydrostatic equilibrium using the same nonrelativistic neutron equation of state.

#### Physical Model

We continue using the nonrelativistic polytropic equation of state derived in Equations (71) to (73):

$$
p=K_{\mathrm{nrel}}\epsilon^{5/3}
$$

The polytropic exponent is:

$$
\gamma=\frac53
$$

and the corresponding polytropic index is:

$$
n=\frac32
$$

The neutron energy density is approximated by its rest mass contribution:

$$
\epsilon\simeq\rho c^2
$$

Unlike the white dwarf model, where electrons provide the degeneracy pressure and nuclei contribute most of the mass, neutrons provide both the pressure and mass energy in this idealised neutron star.

We compare two models.

**1. Newtonian**

The stellar structure equations are Equations (7) and (8):

$$
\frac{dm}{dr}=\frac{4\pi r^2\epsilon}{c^2}
$$

$$
\frac{dp}{dr}=-\frac{G\epsilon m}{c^2r^2}
$$

These equations describe the accumulation of enclosed mass and the decrease in pressure required to balance Newtonian gravity.

**2. TOV**

The enclosed mass equation remains unchanged, but the pressure gradient is replaced by Sagert's Equation (53):

$$
\frac{dp}{dr}
=
-\frac{G\epsilon m}{c^2r^2}
\left(1+\frac{p}{\epsilon}\right)
\left(1+\frac{4\pi r^3p}{mc^2}\right)
\left(1-\frac{2Gm}{rc^2}\right)^{-1}
$$

The three additional factors account for relativistic fluid inertia, the gravitational contribution of pressure and spacetime curvature.

Both models use the same equation of state and central pressures. Therefore, their differences arise from the treatment of gravity rather than the underlying neutron matter model.

For each chosen central pressure, the structure equations are integrated outwards until the pressure reaches zero:

$$
p(R)=0
$$

The stellar radius is $R$, and the total gravitational mass is:

$$
M=m(R)
$$

gives:

$$
{R\propto p_c^{-1/10}}
$$

The predicted radius therefore decreases as central pressure increases.

This is the same mathematical behaviour encountered in the nonrelativistic white dwarf model of Figure 3.

**3. General relativity reduces the predicted stellar mass.**

At the same central pressure, the TOV mass curve lies below the Newtonian mass curve.

The TOV pressure gradient contains three additional correction factors.

For positive pressure and energy density, these factors increase the magnitude of the inward pressure gradient relative to the Newtonian expression for a regular configuration outside its Schwarzschild radius.

Pressure therefore decreases more rapidly with radius.

The resulting equilibrium configurations have lower gravitational masses than their Newtonian counterparts at the same central pressure over the range investigated.

This difference becomes increasingly visible as central pressure increases.

**4. General relativity also reduces the stellar radius.**

The TOV radius curve lies below the Newtonian radius curve.

A stronger relativistic pressure gradient causes pressure to reach zero at a smaller radius.

At:

$$
p_c=10^{33}\ \mathrm{Pa}
$$

the difference between the calculated radii is approximately:

$$
\Delta R=17-16
$$

$$
{\Delta R\approx1\ \mathrm{km}}
$$

The TOV configuration is therefore smaller than the Newtonian configuration calculated using the same central pressure.

The separation between the curves increases with central pressure over the plotted range.

#### Why Is This Result Important?

Figure 8 demonstrates that the predicted properties of neutron stars depend not only on the equation of state but also on the gravitational theory used to calculate hydrostatic equilibrium.

In this calculation, the neutron equation of state is held fixed.

The differences between the Newtonian and TOV curves therefore demonstrate the effects of the general relativistic pressure gradient.

For low central pressures, the corrections are comparatively small, and the two calculations produce similar results.

As central pressure increases, the corrections become more significant.

This is particularly important for neutron stars because their large masses and small radii produce gravitational fields where Newtonian approximations become increasingly inadequate.

Figure 8 also establishes a reference model for the following neutron star calculations.

Figures 9 and 10 will use the full relativistic neutron equation of state rather than the nonrelativistic polytropic approximation.

#### Limitations

The equation of state assumes a zero temperature, noninteracting neutron gas.

The nonrelativistic approximation requires:

$$
k_n\ll m_nc
$$

or equivalently:

$$
{x\ll1}
$$

At higher central densities, neutron momenta increase, and the nonrelativistic approximation becomes progressively less reliable.

Although the TOV calculation includes relativistic gravity, the neutron equation of state remains nonrelativistic.

The model also neglects neutron interactions and additional particle species that become important in realistic neutron star interiors.

Our numerical results show qualitative agreement with Sagert's Figure 8, but a detailed quantitative comparison with the original plotted values has not yet been completed.

Numerical convergence has been examined for the Newtonian model. An equivalent TOV convergence check remains necessary.

The presence of a hydrostatic equilibrium solution also does not, by itself, establish dynamical stability.

#### Conclusion

Figure 8 demonstrates that increasing central pressure produces neutron star configurations with greater masses and smaller radii in both the Newtonian and TOV calculations over the range considered.

The TOV equation predicts lower masses and smaller radii than Newtonian hydrostatic equilibrium for the same central pressure.

The difference grows as relativistic gravitational corrections become more significant.

Because the equation of state is unchanged between the two calculations, this comparison isolates the effects of general relativity on stellar structure.

The result provides the foundation for Figures 9 and 10, where the full relativistic neutron equation of state will be introduced.

Ultrarelativistic Polytropic Case

### Equation (74): Ultrarelativistic Energy Density and Pressure

#### Aim

To derive the ultrarelativistic equation of state for a degenerate neutron gas and investigate why applying this approximation throughout an entire neutron star produces an unphysical stellar model.

Sagert considers the limit:

$$
x\gg1
$$

From Equation (70):

$$
x=\frac{k_n}{m_nc}
$$

Therefore:

$$
k_n\gg m_nc
$$

The neutron Fermi momentum is much greater than the characteristic momentum associated with the neutron rest mass.

We begin with the full energy density and pressure expressions from Equations (68) and (69).

**Energy density:**

$$
\epsilon(x)=\frac{\epsilon_0}{8}
\left[
(2x^3+x)\sqrt{1+x^2}
-\sinh^{-1}(x)
\right]
$$

**Pressure:**

$$
p(x)=\frac{\epsilon_0}{24}
\left[
(2x^3-3x)\sqrt{1+x^2}
+3\sinh^{-1}(x)
\right]
$$

Our objective is to simplify these expressions for $x\gg1$ and eliminate $x$ to obtain a direct relationship between pressure and energy density.

### Ultrarelativistic Energy Density

Consider:

$$
\sqrt{1+x^2}
$$

Factor $x^2$ from the square root:

$$
\sqrt{1+x^2}
=
x\sqrt{1+\frac{1}{x^2}}
$$

Since $x$ is large and positive:

$$
\frac{1}{x^2}\ll1
$$

Therefore:

$$
\sqrt{1+x^2}\approx x
$$

Now consider:

$$
2x^3+x
$$

The cubic term dominates the linear term when $x\gg1$:

$$
2x^3+x\approx2x^3
$$

Combining these approximations:

$$
(2x^3+x)\sqrt{1+x^2}
\approx(2x^3)x
$$

Therefore:

$$
(2x^3+x)\sqrt{1+x^2}\approx2x^4
$$

The remaining term is:

$$
\sinh^{-1}(x)
$$

For large positive $x$:

$$
\sinh^{-1}(x)\approx\ln(2x)
$$

Logarithmic growth is much slower than the growth of $x^4$, so this term becomes negligible relative to the leading term.

Our energy density becomes:

$$
\epsilon(x)\approx\frac{\epsilon_0}{8}(2x^4)
$$

Simplifying:

$$
{
\epsilon(x)\approx\frac{\epsilon_0}{4}x^4
}
$$

This is the first expression in Equation (74).

### Ultrarelativistic Pressure

Return to Equation (69):

$$
p(x)=\frac{\epsilon_0}{24}
\left[
(2x^3-3x)\sqrt{1+x^2}
+3\sinh^{-1}(x)
\right]
$$

Again, in the limit $x\gg1$:

$$
2x^3-3x\approx2x^3
$$

and:

$$
\sqrt{1+x^2}\approx x
$$

Therefore:

$$
(2x^3-3x)\sqrt{1+x^2}
\approx2x^4
$$

The inverse hyperbolic sine term is negligible compared with the leading fourth power.

The pressure expression becomes:

$$
p(x)\approx\frac{\epsilon_0}{24}(2x^4)
$$

Simplifying:

$$
{
p(x)\approx\frac{\epsilon_0}{12}x^4
}
$$

This is the second expression in Equation (74).

### Deriving the Equation of State

We have obtained:

$$
\epsilon=\frac{\epsilon_0}{4}x^4
$$

and:

$$
p=\frac{\epsilon_0}{12}x^4
$$

Both expressions depend on the same dimensionless momentum $x$.

To express pressure directly in terms of energy density, we eliminate $x$ by rearrangement and substitution.

**Step 1: Isolate $x^4$ in the energy density equation.**

Start with:

$$
\epsilon=\frac{\epsilon_0}{4}x^4
$$

Multiply both sides by $4$:

$$
4\epsilon=\epsilon_0x^4
$$

Divide by $\epsilon_0$:

$$
{
x^4=\frac{4\epsilon}{\epsilon_0}
}
$$

**Step 2: Substitute into the pressure equation.**

$$
p=\frac{\epsilon_0}{12}x^4
$$

Substituting our expression for $x^4$:

$$
p=\frac{\epsilon_0}{12}
\left(\frac{4\epsilon}{\epsilon_0}\right)
$$

The factors of $\epsilon_0$ cancel:

$$
p=\frac{4\epsilon}{12}
$$

Simplifying:

$$
{
p=\frac{\epsilon}{3}
}
$$

This is Sagert's Equation (74).

#### Physical Meaning

The pressure is one third of the total energy density in the ultrarelativistic limit.

Both quantities scale with the fourth power of the neutron Fermi momentum:

$$
\epsilon\propto x^4
$$

$$
p\propto x^4
$$

Eliminating this shared dependence produces a linear equation of state:

$$
{p\propto\epsilon}
$$

This contrasts with the nonrelativistic polytropic equation of state derived in Equation (72):

$$
p=K_{\mathrm{nrel}}\epsilon^{5/3}
$$

The distinction arises from the different relationship between particle energy and momentum in the nonrelativistic and ultrarelativistic limits.

### Why Does the Ultrarelativistic Model Fail?

Sagert reports that numerical integration of the TOV equations using Equation (74) does not find a stellar surface.

The normal surface condition is:

$$
p(R)=0
$$

However, in this model, pressure decreases towards zero without reaching it at any finite radius.

Consequently:

$$
p(r)\rightarrow0
\qquad\text{as}\qquad r\rightarrow\infty
$$

The mathematical model therefore has no finite stellar radius.

#### Breakdown of the Approximation

Our pressure equation is:

$$
p(x)\approx\frac{\epsilon_0}{12}x^4
$$

As we move outwards through the star, pressure decreases.

Because $\epsilon_0$ is constant, decreasing pressure corresponds to decreasing $x$:

$$
p\downarrow
\quad\Rightarrow\quad
x\downarrow
$$

Extrapolating the equation towards zero pressure gives:

$$
p\rightarrow0
\quad\Rightarrow\quad
x\rightarrow0
$$

But the equation was derived under the assumption:

$$
x\gg1
$$

As the neutron Fermi momentum decreases, the neutrons eventually enter the nonrelativistic regime.

The ultrarelativistic approximation therefore cannot consistently describe every region of the star.

**The equation of state is valid within its limiting regime. The inconsistency arises from extending that approximation beyond its range of validity.**

### Investigating Sagert's Statement About Stellar Mass

Sagert states that the mass does not become infinite because the mass gradient approaches zero asymptotically.

This statement requires further examination.

From Equation (7), the mass gradient is:

$$
{
\frac{dm}{dr}=\frac{4\pi r^2\epsilon(r)}{c^2}
}
$$

The gradient represents the additional gravitational mass enclosed per unit increase in radius.

Even if the mass gradient approaches zero, the total mass does not necessarily approach a finite limit.

To determine the accumulated mass, we must integrate the gradient.

#### Example 1: A Slowly Decreasing Gradient

Suppose:

$$
\frac{dm}{dr}=\frac1r
$$

The gradient approaches zero as the radius increases:

$$
\frac1r\rightarrow0
\qquad\text{as}\qquad r\rightarrow\infty
$$

The mass accumulated between $r=1$ and $r=R$ is:

$$
m(R)-m(1)=\int_1^R\frac1r\,dr
$$

Integrating:

$$
=\left[\ln r\right]_1^R
$$

Applying the limits:

$$
=\ln R-\ln1
$$

Since $\ln1=0$:

$$
\boxed{
m(R)-m(1)=\ln R
}
$$

Although logarithmic growth becomes progressively slower:

$$
\lim_{R\to\infty}\ln R=\infty
$$

Therefore, the total accumulated mass grows without limit even though its gradient approaches zero.

#### Example 2: A More Rapidly Decreasing Gradient

Now suppose:

$$
\frac{dm}{dr}=\frac1{r^2}
$$

Again:

$$
\frac{dm}{dr}\rightarrow0
$$

The accumulated mass is:

$$
m(R)-m(1)=\int_1^Rr^{-2}\,dr
$$

Using the power rule:

$$
\int r^{-2}\,dr=-r^{-1}+C
$$

Therefore:

$$
m(R)-m(1)=\left[-\frac1r\right]_1^R
$$

Applying the limits:

$$
=-\frac1R-\left(-1\right)
$$

Hence:

$$
{
m(R)-m(1)=1-\frac1R
}
$$

As $R\rightarrow\infty$:

$$
{
m(R)-m(1)\rightarrow1
}
$$

The accumulated mass is finite.

#### Interpretation

Both examples have mass gradients that approach zero.

However, the first produces an unbounded accumulated mass, while the second produces a finite accumulated mass.

Therefore:

$$
{
\frac{dm}{dr}\rightarrow0
\quad\not\Rightarrow\quad
m(r)\rightarrow\text{finite limit}
}
$$

Knowing that the mass gradient decreases is insufficient.

We need to establish how rapidly it decreases with radius.

### Investigating the Actual TOV Mass Gradient

Our mass equation is:

$$
\frac{dm}{dr}=\frac{4\pi r^2\epsilon(r)}{c^2}
$$

There are two competing radial effects.

The factor $r^2$ increases with radius.

The energy density $\epsilon(r)$ decreases with radius.

We therefore need to determine the radial dependence of energy density.

#### Proposed Large Radius Behaviour

We investigate the possibility that:

$$
{
\epsilon(r)\sim\frac{A}{r^2}
}
$$

where $A$ is a positive constant.

This is initially a proposed asymptotic behaviour, not an established result.

Substitute into Equation (7):

$$
\frac{dm}{dr}
=
\frac{4\pi r^2}{c^2}
\left(\frac{A}{r^2}\right)
$$

The factors of $r^2$ cancel:

$$
{
\frac{dm}{dr}\sim\frac{4\pi A}{c^2}
}
$$

The mass gradient therefore approaches a positive constant if this proposed energy density is correct.

Integrating the corresponding leading behaviour gives:

$$
m(r)\sim\frac{4\pi A}{c^2}r
$$

Thus:

$$
{m(r)\propto r}
$$

In this case, mass increases proportionally to radius and does not approach a finite limit.

However, we must first test whether our proposed energy density is consistent with the TOV equation.

### Testing the Proposed Energy Density Against TOV

The TOV pressure equation is:

$$
\frac{dp}{dr}
=
-\frac{G}{r^2}
\left(\frac{\epsilon+p}{c^2}\right)
\left(m+\frac{4\pi r^3p}{c^2}\right)
\left(1-\frac{2Gm}{rc^2}\right)^{-1}
$$

We will substitute our proposed asymptotic energy density:

$$
\epsilon(r)\sim\frac{A}{r^2}
$$

and determine whether there is a positive value of $A$ satisfying the equation at large radii.

#### Step 1: Calculate Pressure

From Equation (74):

$$
p=\frac{\epsilon}{3}
$$

Substitute the proposed energy density:

$$
{
p(r)\sim\frac{A}{3r^2}
}
$$

#### Calculate the Pressure Gradient

Differentiate:

$$
p(r)=\frac{A}{3}r^{-2}
$$

Using the power rule:

$$
\frac{dp}{dr}
=
\frac{A}{3}(-2r^{-3})
$$

Therefore:

$$
{
\frac{dp}{dr}\sim-\frac{2A}{3r^3}
}
$$

This is the pressure gradient predicted by our proposed solution.

We must now compare it with the TOV equation.

#### Simplify the First TOV Bracket

Consider:

$$
\frac{\epsilon+p}{c^2}
$$

Substituting our energy density and pressure:

$$
\frac{\epsilon+p}{c^2}
=
\frac1{c^2}
\left(
\frac{A}{r^2}+\frac{A}{3r^2}
\right)
$$

Using a common denominator:

$$
\frac{A}{r^2}+\frac{A}{3r^2}
=
\frac{3A+A}{3r^2}
$$

Therefore:

$$
{
\frac{\epsilon+p}{c^2}
=
\frac{4A}{3r^2c^2}
}
$$

#### Simplify the Second TOV Bracket

Consider:

$$
m+\frac{4\pi r^3p}{c^2}
$$

We found from the mass equation:

$$
m(r)=\frac{4\pi Ar}{c^2}+C
$$

where $C$ is an integration constant within the proposed radial solution.

We also know:

$$
p(r)=\frac{A}{3r^2}
$$

Substitute into the pressure contribution:

$$
\frac{4\pi r^3p}{c^2}
=
\frac{4\pi r^3}{c^2}
\left(\frac{A}{3r^2}\right)
$$

Cancelling the powers of $r$:

$$
{
\frac{4\pi r^3p}{c^2}
=
\frac{4\pi Ar}{3c^2}
}
$$

Now combine this with the mass:

$$
m+\frac{4\pi r^3p}{c^2}
=
\frac{4\pi Ar}{c^2}
+
\frac{4\pi Ar}{3c^2}
+C
$$

Using a common denominator:

$$
=
\frac{12\pi Ar}{3c^2}
+
\frac{4\pi Ar}{3c^2}
+C
$$

Adding like terms:

$$
{
m+\frac{4\pi r^3p}{c^2}
=
\frac{16\pi Ar}{3c^2}+C
}
$$

At large radii, the term proportional to $r$ dominates the constant $C$.

Therefore:

$$
{
m+\frac{4\pi r^3p}{c^2}
\sim\frac{16\pi Ar}{3c^2}
}
$$

#### Simplify the Third TOV Bracket

The final factor is:

$$
\left(1-\frac{2Gm}{rc^2}\right)^{-1}
$$

Substitute:

$$
m(r)=\frac{4\pi Ar}{c^2}+C
$$

Therefore:

$$
\left[
1-\frac{2G}{rc^2}
\left(\frac{4\pi Ar}{c^2}+C\right)
\right]^{-1}
$$

Distribute the factor:

$$
=
\left(
1-\frac{8\pi GA}{c^4}-\frac{2GC}{rc^2}
\right)^{-1}
$$

At large radii:

$$
\frac{2GC}{rc^2}\rightarrow0
$$

So the final TOV factor approaches:

$$
\left(1-\frac{8\pi GA}{c^4}\right)^{-1}
$$

A negative exponent means taking the reciprocal:

$$
a^{-1}=\frac1a
$$

Therefore:

$$
=
\frac1{1-\frac{8\pi GA}{c^4}}
$$

Writing the denominator as a single fraction:

$$
1-\frac{8\pi GA}{c^4}
=
\frac{c^4-8\pi GA}{c^4}
$$

Taking the reciprocal gives:

$$
{
\left(1-\frac{8\pi GA}{c^4}\right)^{-1}
=
\frac{c^4}{c^4-8\pi GA}
}
$$

#### Substitute Everything into TOV

The proposed pressure gradient is:

$$
\frac{dp}{dr}\sim-\frac{2A}{3r^3}
$$

The right side of the TOV equation becomes:

$$
-\frac{G}{r^2}
\left(\frac{4A}{3r^2c^2}\right)
\left(\frac{16\pi Ar}{3c^2}\right)
\left(\frac{c^4}{c^4-8\pi GA}\right)
$$

First, consider the powers of radius:

$$
\frac1{r^2}\frac1{r^2}r
=
\frac{r}{r^4}
=
\frac1{r^3}
$$

Next, combine the numerical factors:

$$
\frac43\times\frac{16}{3}
=
\frac{64}{9}
$$

The two factors of $A$ give:

$$
A\times A=A^2
$$

The factors of $c^4$ cancel.

Therefore:

$$
{
-\frac{2A}{3r^3}
=
-\frac{64\pi GA^2}
{9r^3(c^4-8\pi GA)}
}
$$

Both sides have the same radial dependence.

We can now investigate whether their coefficients agree.

#### Determine the Constant A

Starting with:

$$
-\frac{2A}{3r^3}
=
-\frac{64\pi GA^2}
{9r^3(c^4-8\pi GA)}
$$

Cancel the negative signs and the common factors of $r^{-3}$:

$$
\frac{2A}{3}
=
\frac{64\pi GA^2}
{9(c^4-8\pi GA)}
$$

Multiply both sides by $3$:

$$
2A=
\frac{64\pi GA^2}
{3(c^4-8\pi GA)}
$$

Divide both sides by $2$:

$$
A=
\frac{64\pi GA^2}
{6(c^4-8\pi GA)}
$$

Assuming $A\neq0$, divide both sides by $A$:

$$
1=
\frac{64\pi GA}
{6(c^4-8\pi GA)}
$$

Multiply both sides by the denominator:

$$
6(c^4-8\pi GA)=64\pi GA
$$

Expand:

$$
6c^4-48\pi GA=64\pi GA
$$

Add $48\pi GA$ to both sides:

$$
6c^4=112\pi GA
$$

Divide both sides by $112\pi G$:

$$
A=\frac{6c^4}{112\pi G}
$$

Simplify:

$$
{
A=\frac{3c^4}{56\pi G}
}
$$

We have found a positive, nonzero constant that makes our proposed large radius behaviour consistent with the TOV equations.

Therefore:

$$
{
\epsilon(r)\sim\frac{3c^4}{56\pi Gr^2}
}
$$

The corresponding pressure is:

$$
{
p(r)\sim\frac{c^4}{56\pi Gr^2}
}
$$

This establishes a consistent asymptotic solution of the TOV equations for the equation of state $p=\epsilon/3$.

It does not, by itself, establish that a stellar solution starting from regular central conditions approaches this asymptotic behaviour.

### Consequences for the Enclosed Mass

Recall Equation (7):

$$
\frac{dm}{dr}=\frac{4\pi r^2\epsilon(r)}{c^2}
$$

Substitute our asymptotic energy density:

$$
\frac{dm}{dr}
\sim
\frac{4\pi r^2}{c^2}
\left(\frac{3c^4}{56\pi Gr^2}\right)
$$

Cancel the factors of $r^2$ and $\pi$:

$$
\frac{dm}{dr}
\sim
\frac{12c^4}{56Gc^2}
$$

Simplifying:

$$
{
\frac{dm}{dr}\rightarrow\frac{3c^2}{14G}
}
$$

The mass gradient approaches a positive constant rather than zero.

Integrating its leading behaviour:

$$
m(r)\sim\int\frac{3c^2}{14G}\,dr
$$

Therefore:

$$
{
m(r)\sim\frac{3c^2}{14G}r
}
$$

The enclosed mass grows approximately linearly with radius.

Consequently:

$$
{
r\rightarrow\infty
\quad\Rightarrow\quad
m(r)\rightarrow\infty
}
$$

This result disagrees with Sagert's statement that the mass remains finite.

However, we still need to establish whether a numerical solution beginning at the stellar centre approaches the asymptotic solution.

### Independent Numerical Verification

A separate numerical integration of the TOV equations was performed using:

$$
p=\frac{\epsilon}{3}
$$

The calculation used a positive central pressure and regular central boundary conditions.

The equations were expressed in dimensionless quantities to avoid working directly with extremely large physical numbers.

Define:

$$
s=\frac{r}{L}
$$

$$
q=\frac{\epsilon}{\epsilon_c}
$$

$$
\mu=\frac{Gm}{c^2L}
$$

where $\epsilon_c$ is the central energy density and $L$ is the length scale defined by:

$$
L^2=\frac{c^4}{4\pi G\epsilon_c}

The resulting numerical solution was examined at progressively larger dimensionless radii.

| Scaled radius $s$ | $s^2q$ | $\mu/s$ |
|---|---|---|
| $100$ | $0.21568$ | $0.21641$ |
| $1,000$ | $0.21457$ | $0.21396$ |
| $10,000$ | $0.21417$ | $0.21430$ |
| $100,000$ | $0.21430$ | $0.21429$ |
| $1,000,000$ | $0.21428$ | $0.21428$ |

Both quantities approach approximately:

$$
\frac3{14}\approx0.214286
$$

#### Interpretation of the Energy Density

The results show:

$$
s^2q\rightarrow\frac3{14}
$$

Rearranging:

$$
\boxed{
q\sim\frac{3}{14s^2}
}

Since $q$ is proportional to energy density and $s$ is proportional to radius:

$$
\boxed{
\epsilon(r)\propto\frac1{r^2}

This agrees with our proposed asymptotic energy density.

#### Interpretation of the Mass

The results also show:

\frac{\mu}{s}\rightarrow\frac3{14}

Rearranging:

$$
\boxed{
\mu\sim\frac3{14}s
}

Since $\mu$ is proportional to enclosed mass and $s$ is proportional to radius:

$$
\boxed{
m(r)\propto r
}
$$

The numerical results therefore support the asymptotic behaviour derived analytically.

In particular, they show that the regular stellar solution approaches a regime in which the enclosed mass grows approximately linearly with radius.

### Comparison with Sagert's Statement

Sagert correctly identifies that the ultrarelativistic equation of state:

$$
p=\frac{\epsilon}{3}
$$

does not produce a finite stellar surface when applied throughout the star.

Pressure approaches zero asymptotically rather than reaching zero at a finite radius.

However, Sagert additionally states that the mass remains finite because the mass gradient approaches zero.

Our investigation gives a different result.


$$
\epsilon(r)\sim\frac{3c^4}{56\pi Gr^2}
$$

$$
\frac{dm}{dr}\rightarrow\frac{3c^2}{14G}
$$

$$
m(r)\sim\frac{3c^2}{14G}r
$$


The integrated TOV solution approaches the same radial behaviour.

Therefore, the mathematical model described by the equation of state $p=\epsilon/3$ has both an unbounded radius and an unbounded enclosed mass.

The asymptotic mass gradient approaches a positive constant, not zero.

This indicates an error in Sagert's statement about the asymptotic mass behaviour.

The correction does not change the paper's main physical conclusion that the ultrarelativistic approximation is unsuitable for describing an entire neutron star.

Equation (74) describes the ultrarelativistic limit of a degenerate neutron gas:

$$
{
p=\frac{\epsilon}{3}
}
$$

This equation of state is valid when the neutron Fermi momentum is much greater than the neutron rest mass momentum.

However, as pressure decreases towards the outer regions of the star, the neutron Fermi momentum also decreases.

The system necessarily leaves the ultrarelativistic regime before reaching zero pressure.

Applying the limiting equation of state throughout the star is therefore physically inconsistent.

The resulting TOV solution has no finite stellar surface.

Our additional analytical and numerical investigation also shows that its enclosed mass grows without limit.
