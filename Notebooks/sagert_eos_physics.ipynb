## Compact stars as stellar remnants

Compact stars, particularly white dwarfs and neutron stars, are final stages in the evolution of ordinary stars.

After hydrogen and helium burning, helium in the stellar core is converted mainly into carbon and oxygen. When nuclear burning in the core stops, the temperature decreases and thermal support is reduced.

Gravity then causes the star to contract. As the same stellar material is compressed into a smaller volume, the density and pressure in the core increase.

If the star is sufficiently massive, gravitational contraction can raise the core temperature and density enough to begin further fusion reactions involving heavier elements.

For stars with an initial mass below approximately $8M_\odot$ the gravitational compression is not sufficient to reach the conditions required for carbon fusion.

The hot outer regions expand and can be lost through stellar winds, leaving behind a compact core composed mainly of carbon and oxygen.

This compact remnant becomes a white dwarf.

### White dwarfs

A white dwarf no longer contains ongoing fusion capable of supplying the thermal pressure required to support an ordinary star.

A different source of pressure is therefore required to maintain hydrostatic equilibrium.

This pressure is provided by a degenerate electron gas.

### Degenerate matter

Fermions obey the Pauli exclusion principle, so they cannot occupy the same quantum state. As the lowest momentum states become occupied, additional fermions must occupy progressively higher momentum states.

Compressing the gas therefore forces the occupied states to extend to higher momenta. This produces degeneracy pressure, which resists further compression even at very low temperature.

In a white dwarf, the nuclei contain most of the mass, while the degenerate electrons provide most of the pressure supporting the star. Unlike thermal pressure maintained by nuclear fusion, electron degeneracy pressure does not require an ongoing source of thermal energy.

If gravity compresses the star further, the available low momentum states become increasingly occupied. The electrons must then occupy higher momentum states, increasing the degeneracy pressure and resisting further gravitational compression.

#### Pauli exclusion principle

Electrons are fermions. The Pauli exclusion principle states that two identical fermions cannot occupy the same quantum state.

As more electrons are compressed into a given volume, the available low momentum states become occupied. Additional electrons must therefore occupy states with progressively higher momentum.

The resulting increase in electron momentum produces **degeneracy pressure**, which acts against gravitational compression and helps stabilise the white dwarf.

The balance is therefore

gravity pushes inward, electron degeneracy pressure pushes outward

#### Why electrons become degenerate first

Electrons become degenerate before the much heavier nuclei because of their small mass. Their characteristic quantum wavelength is larger, so their wavefunctions begin to overlap at lower densities.

Once this overlap becomes important, the electrons must be treated as a quantum system governed by Fermi statistics and the Pauli exclusion principle.

#### Increasing white dwarf mass

White dwarfs contain nuclei, mainly carbon and oxygen, but in the density range considered here their contribution to the pressure is negligible. The pressure support is dominated by the degenerate electrons.

As the mass of the white dwarf increases, its gravitational force becomes stronger. A greater electron degeneracy pressure is therefore required to maintain stability, which requires the matter to reach a greater density.

There is a maximum mass beyond which electron degeneracy pressure can no longer prevent gravitational collapse. This is the Chandrasekhar mass, approximately

$$
M_{\mathrm{Ch}} \approx 1.4M_\odot
$$

White dwarfs have typical radii of approximately

$$
R_{\mathrm{WD}} \sim 10{,}000\ \mathrm{km}
$$

For neutron stars, the corresponding degeneracy pressure originates mainly from neutrons rather than electrons.

### Formation of neutron star matter

For stars with initial masses greater than approximately $8M_\odot$, fusion can continue beyond carbon and oxygen to produce progressively heavier elements.

Once iron is produced in the core, further fusion no longer provides energy to support the star. As thermal support decreases, the iron core contracts under gravity. Its density increases as its radius decreases and its mass continues to grow.

At sufficiently high density, electrons are captured by nuclei and combine with protons to form neutrons:

$$
p + e^- \rightarrow n.
$$

The nuclei therefore become increasingly neutron rich.

As the density increases further, the nuclei eventually become unable to bind all of their neutrons. Neutrons then begin to leave the nuclei and form a neutron liquid surrounding them. This is neutron drip.

Importantly, the nuclei have not disappeared at neutron drip. The matter consists of neutron rich nuclei together with free neutrons.

At still greater densities, the matter becomes increasingly neutron dominated, with a small fraction of protons and electrons.

Because neutrons are fermions, the Pauli exclusion principle becomes important as they are packed into increasingly dense matter. The neutrons become degenerate in the same way that electrons become degenerate in white dwarfs.

The progression described in the paper is therefore:

$$
\text{neutron rich nuclei}
\rightarrow
\text{neutron drip}
\rightarrow
\text{nuclei + free neutron liquid}
\rightarrow
\text{neutron dominated matter}.
$$

### Core collapse and the neutron star remnant

As the stellar core collapses, increasingly dense neutron rich matter is formed. Sagert describes the high incompressibility of neutron star matter as producing an outgoing shock wave.

The shock interacts with the still collapsing outer layers of the star, producing the outward expansion associated with a core collapse supernova.

The neutron rich remnant can become a neutron star. Sagert states that remnants with masses greater than approximately $2$ to $3M_\odot$ instead continue to collapse towards a black hole.

A typical neutron star has a radius of approximately

$$
R_{\mathrm{NS}} \sim 10\ \mathrm{km},
$$

compared with approximately

$$
R_{\mathrm{WD}} \sim 10{,}000\ \mathrm{km}
$$

for a white dwarf.

For a spherical object,

$$
V=\frac{4}{3}\pi R^3.
$$

Therefore, for fixed mass,

$$
\rho \propto \frac{1}{R^3}.
$$

The cubic dependence arises because volume extends in three spatial dimensions. A reduction in radius therefore produces a much larger change in density.

### Structure equations

A compact star can be treated as a series of thin spherical shells. At a radius $r$, the quantity $m(r)$ represents all of the mass enclosed within that radius, while $dm$ represents the mass contained in a thin shell of thickness $dr$.

Pressure is force per unit area. For a spherical surface at radius $r$,

$$
dp = \frac{dF}{4\pi r^2}.
\tag{1}
$$

The gravitational force acting on a thin shell of mass $dm$ is

$$
dF = -\frac{G\,dm\,m(r)}{r^2}.
\tag{2}
$$

The negative sign appears because the radial coordinate increases outwards, while gravity acts inwards towards the centre of the star.

The mass of the thin shell is its local density multiplied by its volume:

$$
dm = \rho(r)dV.
$$

For a spherical shell with surface area $4\pi r^2$ and thickness $dr$,

$$
dV = 4\pi r^2dr,
$$

so

$$
dm = \rho(r)4\pi r^2dr.
\tag{3}
$$

Substituting equations (2) and (3) into equation (1) gives

$$
dp
=
-\frac{G\rho(r)m(r)}{r^2}dr.
$$

Dividing by $dr$ gives the pressure structure equation,

$$
\frac{dp}{dr}
=
-\frac{G\rho(r)m(r)}{r^2}.
\tag{4}
$$

The negative derivative means that pressure decreases as the radius increases.

Equation (3) can also be divided by $dr$ to give

$$
\frac{dm}{dr}
=
4\pi r^2\rho(r).
\tag{5}
$$

This derivative is positive because increasing the radius encloses progressively more of the star's mass.

Sagert expresses the mass density in terms of the energy density $\epsilon(r)$ using

$$
\rho(r)=\frac{\epsilon(r)}{c^2},
\tag{6}
$$

which is the mass energy relation expressed per unit volume.

Substituting this into equations (4) and (5) gives

$$
\frac{dm}{dr}
=
\frac{4\pi r^2\epsilon(r)}{c^2},
\tag{7}
$$

and

$$
\frac{dp}{dr}
=
-\frac{G\epsilon(r)m(r)}{c^2r^2}.
\tag{8}
$$

These are coupled differential equations: the quantities describing the star must be evolved together as the radius changes.

At the centre of the star,

$$
m(0)=0,
$$

because a sphere of zero radius encloses no volume and therefore no mass. The central pressure is specified as

$$
p(0)=p_0.
$$

Moving outwards from the centre,

$$
m(r)\uparrow
\qquad\text{while}\qquad
p(r)\downarrow.
$$

The stellar surface is reached when the pressure falls to zero.

Equations (7) and (8) are not sufficient by themselves because they also contain the energy density $\epsilon(r)$. A relation between pressure and energy density is therefore required

$$
p=p(\epsilon)
$$

or equivalently

$$
\epsilon=\epsilon(p).
$$

This relation is provided by the equation of state.

### Fermi Dirac distribution

The occupation of electron energy states is described by the Fermi Dirac distribution,

$$
f(E)=\frac{1}{\exp[(E-\mu)/k_BT]+1}.
\tag{9}
$$

Here,

- $f(E)$ is the probability that a state with energy $E$ is occupied,
- $E$ is the energy of the state,
- $\mu$ is the chemical potential,
- $k_B$ is the Boltzmann constant,
- $T$ is the temperature.

The chemical potential provides the characteristic energy around which the occupation of states changes from likely occupied to likely empty.

When

$$
E=\mu,
$$

the exponent is zero, giving

$$
f(E)=\frac{1}{e^0+1}=\frac{1}{2}.
$$

States with energies well below $\mu$ have occupation probabilities close to one,

$$
E\ll\mu,\qquad\ then \qquad f(E)\approx1,
$$

while states with energies well above $\mu$ have occupation probabilities close to zero,

$$
E\gg\mu,\qquad\ then \qquad f(E)\approx0.
$$

Temperature controls how sharply the occupation probability changes around $\mu$. At higher temperatures the transition is more spread out, while at very low temperatures it becomes increasingly sharp.

### Momentum space

At zero temperature, electrons occupy the available momentum states from zero momentum up to the Fermi momentum, $k_F$. In three dimensional momentum space, these occupied states form a sphere with radius $k_F$, known as the Fermi sphere.

The volume of occupied momentum space therefore scales as $k_F^3$. Since increasing the electron number density requires more momentum states to be occupied, the electron number density also scales as $k_F^3$.

#### Phase space and quantum states

A particle can be described by both its position and its momentum.

In three dimensions, its position is described by

$$
(x,y,z),
$$

while its momentum is described by

$$
(p_x,p_y,p_z).
$$

Together these form a six dimensional **phase space**,

$$
(x,y,z,p_x,p_y,p_z).
$$

Two particles can therefore have the same position but occupy different points in phase space if their momenta are different. Similarly, particles with the same momentum can occupy different points in phase space if their positions are different.

Quantum mechanics restricts the allowed momentum states of a particle confined within a finite region. To see how this leads to the phase space cell used by Sagert et al., consider a one dimensional box of length $L$ with periodic boundary conditions.

The allowed wavelengths satisfy

$$
L=n\lambda,
$$

so

$$
\lambda_n=\frac{L}{n}.
$$

Using the de Broglie relation,

$$
p=\frac{h}{\lambda},
$$

the corresponding allowed momenta are

$$
p_n=\frac{h}{\lambda_n}
=\frac{nh}{L}.
$$

The spacing between neighbouring momentum states is therefore

$$
\Delta p
=
p_{n+1}-p_n
=
\frac{(n+1)h}{L}-\frac{nh}{L}
=
\frac{h}{L}.
$$

Since

$$
h=2\pi\hbar,
$$

this becomes

$$
\Delta p=\frac{2\pi\hbar}{L}.
$$

In three dimensions,

$$
\Delta p_x
=
\Delta p_y
=
\Delta p_z
=
\frac{2\pi\hbar}{L}.
$$

The volume occupied by one state in momentum space is therefore

$$
\Delta p_x\Delta p_y\Delta p_z
=
\left(\frac{2\pi\hbar}{L}\right)^3.
$$

Since the physical volume of the box is

$$
V=L^3,
$$

this can be written as

$$
\Delta p_x\Delta p_y\Delta p_z
=
\frac{(2\pi\hbar)^3}{V}.
$$

Multiplying by the corresponding position space volume gives the phase space volume

$$
V\Delta p_x\Delta p_y\Delta p_z
=
(2\pi\hbar)^3.
$$

This is the phase space cell volume used by Sagert et al. when counting the available quantum states in Eq. (10).

#### Counting occupied states

Sagert et al. relate the number of particles to the number of available quantum states using

$$
\frac{dn_i}{d^3k}
=
\frac{g}{(2\pi\hbar)^3}f.
\tag{10}
$$

Here,

- $dn_i$ is a small contribution to the number density of particles of species $i$,
- $d^3k$ is a small volume in three dimensional momentum space,
- $g$ is the number of internal states available for a given momentum,
- $f$ is the occupation probability given by the Fermi Dirac distribution.

For electrons there are two possible spin states,

$$
m_s=+\frac{1}{2}
\qquad\text{and}\qquad
m_s=-\frac{1}{2},
$$

so

$$
g=2.
$$

The factor

$$
\frac{1}{(2\pi\hbar)^3}
$$

gives the density of available quantum states in phase space. Multiplying by $g$ accounts for the possible spin states, while multiplying by $f$ accounts for the probability that those states are occupied.

Therefore,

$$
\frac{dn_i}{d^3k}
$$

describes the particle number density per unit volume of momentum space.

Conceptually, Eq. (10) can be read as

$$
\text{occupied state density}
=
\text{available state density}
\times
\text{occupation probability}.
$$

#### Total particle number density

Equation (10) gives the particle number density per unit volume of momentum space. To obtain the total number density of particles of species $i$, these contributions are added over momentum space

$$
n_i
=
\int dn_i
=
\int
\frac{g}{(2\pi\hbar)^3}f\,d^3k.
\tag{11}
$$

Here,

$$
d^3k=dk_x\,dk_y\,dk_z
$$

represents a small volume in three dimensional momentum space.

The quantity

$$
\frac{g}{(2\pi\hbar)^3}f\,d^3k
$$

therefore gives the contribution to the particle number density from states within a small region of momentum space.

Integration adds these small contributions together:

$$
n_i=\int dn_i.
$$

Equation (11) can therefore be interpreted as adding the contributions from particles occupying states throughout momentum space to obtain the total particle number density $n_i$.

#### Zero temperature limit and the Fermi energy

For a strongly degenerate electron gas, Sagert et al. assume that the energy scale associated with degeneracy is much greater than the thermal energy scale:

$$
\mu-m_ec^2 \gg k_BT.
$$

Here, $m_ec^2$ is the rest mass energy of an electron. Since thermal effects are small compared with the degeneracy energy scale, the electron gas can be approximated as having

$$
T=0.
$$

This does not mean that the physical temperature of the white dwarf is literally absolute zero. It means that thermal effects can be neglected in the model.

In this limit, the chemical potential is identified with the Fermi energy,

$$
\mu=E_F.
$$

The Fermi Dirac distribution then becomes

$$
f(E)=
\begin{cases}
1, & E\leq E_F,\\
0, & E>E_F.
\end{cases}
\tag{12}
$$

All states with energies up to the Fermi energy are therefore occupied, while states above the Fermi energy are empty.

The occupied states form the Fermi sea. The boundary of the occupied region is the Fermi surface. The energy at this boundary is the Fermi energy, $E_F$, and the corresponding momentum is the Fermi momentum, $k_F$.

#### Electron number density and the Fermi sphere

At zero temperature, all electron states up to the Fermi momentum $k_F$ are occupied. Equation (11) can therefore be applied with

$$
f=1
$$

throughout the occupied region of momentum space.

Electrons have two possible spin states, so $g=2$. The electron number density is therefore

$$
n_e
=
\int_0^{k_F}
\frac{2}{(2\pi\hbar)^3}\,d^3k.
$$

The occupied states form a sphere in three dimensional momentum space with radius $k_F$, known as the **Fermi sphere**.

A thin spherical shell in momentum space has volume

$$
d^3k=4\pi k^2\,dk.
$$

This has the same geometrical form as a thin shell in physical space, $dV=4\pi r^2dr$, but here the radius $k$ is a momentum coordinate rather than a physical distance.

Substituting the momentum space volume element gives

$$
n_e
=
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_F}k^2\,dk.
$$

Using

$$
\int_0^{k_F}k^2\,dk
=
\left[\frac{k^3}{3}\right]_0^{k_F}
=
\frac{k_F^3}{3},
$$

gives

$$
n_e
=
\frac{8\pi}{(2\pi\hbar)^3}
\frac{k_F^3}{3}.
$$

Since

$$
(2\pi\hbar)^3=8\pi^3\hbar^3,
$$

this simplifies to

$$
{
n_e=\frac{k_F^3}{3\pi^2\hbar^3}
}.
\tag{13}
$$

The electron number density therefore scales as

$$
n_e\propto k_F^3.
$$

The cubic dependence arises because the occupied states fill a three dimensional region of momentum space. For example, doubling the Fermi momentum increases the occupied momentum space volume, and therefore the electron number density, by a factor of

$$
(2k_F)^3 = 2^3k_F^3 = 8k_F^3.
$$

#### Mass density of white dwarf matter

White dwarf matter must be electrically neutral, so the number of electrons is equal to the number of protons contained in the nuclei.

Sagert et al. consider white dwarfs composed predominantly of carbon or oxygen. For a nucleus,

- $A$ is the total number of nucleons, including protons and neutrons,
- $Z$ is the number of protons.

For both carbon 12 and oxygen 16,

$$
\frac{A}{Z}=2.
$$

There are therefore two nucleons associated with each electron in electrically neutral carbon or oxygen matter.

Since protons and neutrons are much more massive than electrons, Sagert et al. neglect the electron mass when calculating the mass density of the white dwarf. Using $m_N$ for the nucleon mass, the mass density is

$$
{
\rho=n\,m_N\frac{A}{Z}
}.
\tag{14}
$$

Here, $n=n_e$ is the electron number density.

The factors can be interpreted as

$$
\underbrace{n}_{\text{electrons per volume}}
\underbrace{\frac{A}{Z}}_{\text{nucleons per electron}}
\underbrace{m_N}_{\text{mass per nucleon}}
=
\underbrace{\rho}_{\text{mass per volume}}.
$$

For a fixed composition, $m_N$ and $A/Z$ are constant, so

$$
\rho\propto n.
$$

The nuclei therefore provide almost all of the white dwarf's mass density, while the degenerate electrons will provide the dominant pressure.

#### Fermi momentum and mass density

Equation (13) relates the electron number density to the Fermi momentum,

$$
n=\frac{k_F^3}{3\pi^2\hbar^3},
$$

while equation (14) relates the electron number density to the mass density,

$$
\rho=n\,m_N\frac{A}{Z}.
$$

Rearranging equation (14) for $n$ gives

$$
n=\frac{\rho Z}{m_NA}.
$$

Since both expressions describe the same number density, they can be equated:

$$
\frac{\rho Z}{m_NA}
=
\frac{k_F^3}{3\pi^2\hbar^3}.
$$

Multiplying both sides by $3\pi^2\hbar^3$ gives

$$
k_F^3
=
3\pi^2\hbar^3
\left(
\frac{\rho Z}{m_NA}
\right).
$$

Taking the cube root of both sides,

$$
k_F
=
\left[
3\pi^2\hbar^3
\left(
\frac{\rho Z}{m_NA}
\right)
\right]^{1/3}.
$$

Since

$$
(\hbar^3)^{1/3}=\hbar,
$$

this becomes

$$
{
k_F
=
\hbar
\left(
\frac{3\pi^2\rho Z}{m_NA}
\right)^{1/3}
}.
\tag{15}
$$

For a fixed composition,

$$
k_F\propto\rho^{1/3}.
$$

Increasing the mass density therefore increases the Fermi momentum. Physically, a greater density means that more electrons are packed into a given volume, so progressively higher momentum states must be occupied.

The cube root reflects the three dimensional nature of momentum space.

For example, if the mass density becomes

$$
8\rho,
$$

then the Fermi momentum becomes

$$
2k_F.
$$

An eightfold increase in mass density therefore produces a twofold increase in Fermi momentum.

#### Energy density of white dwarf matter

The total energy density of the white dwarf contains contributions from both the nuclei and the electrons.

Sagert et al. write

$$
{
\epsilon
=
n m_N\frac{A}{Z}c^2
+
\epsilon_{\mathrm{elec}}(k_F)
}.
\tag{16}
$$

From equation (14),

$$
\rho=n m_N\frac{A}{Z},
$$

so the first term can be written as

$$
n m_N\frac{A}{Z}c^2
=
\rho c^2.
$$

This represents the rest mass energy density of the nuclei.

The nuclei contain almost all of the mass of the white dwarf because protons and neutrons are much more massive than electrons. Their contribution therefore dominates the total energy density.

The second term,

$$
\epsilon_{\mathrm{elec}}(k_F),
$$

is the energy density of the degenerate electrons. Its dependence on $k_F$ reflects the energies of the occupied electron momentum states.

Although the electrons contribute relatively little to the mass and total energy density, they provide most of the pressure supporting the white dwarf.

The roles of the two components are therefore different:

$$
\text{nuclei = most of the mass and energy density},
$$

while

$$
\text{electrons = most of the pressure}.
$$

#### Relativistic electron energy

The energy of an electron with momentum $k$ is given by the relativistic energy momentum relation,

$$
\boxed{
E(k)=\sqrt{k^2c^2+m_e^2c^4}
}.
\tag{17}
$$

Here,

- $E(k)$ is the energy of an electron with momentum $k$,
- $k$ is the electron momentum,
- $m_e$ is the electron mass,
- $c$ is the speed of light.

If the electron has zero momentum,

$$
k=0,
$$

then

$$
E(0)
=
\sqrt{m_e^2c^4}
=
m_ec^2.
$$

The electron therefore still possesses its rest mass energy even when its momentum is zero.

At very large momentum, the momentum term becomes much larger than the rest mass term. The energy then approaches

$$
E(k)\approx\sqrt{k^2c^2}=kc.
$$

This is the relativistic high momentum limit.

As the density of the white dwarf increases, the Fermi momentum $k_F$ increases. The energy of the highest occupied electron states therefore also increases. At sufficiently high momentum, the electrons become relativistic, although their speed approaches rather than reaches the speed of light.

### Electron energy density: evaluating the integral

After introducing the dimensionless momentum

$$
u=\frac{k}{m_ec},
$$

the electron energy density becomes

$$
\epsilon_{\rm elec}(k_F)
=
\epsilon_0
\int_0^x u^2\sqrt{1+u^2}\,du,
$$

where

$$
x=\frac{k_F}{m_ec},
$$

and

$$
\epsilon_0=
\frac{m_e^4c^5}{\pi^2\hbar^3}.
$$

The remaining integral is

$$
\int u^2\sqrt{1+u^2}\,du.
$$

Because the integrand contains $\sqrt{1+u^2}$, use the hyperbolic substitution

$$
u=\sinh\theta.
$$

Then

$$
du=\cosh\theta\,d\theta,
$$

and using

$$
1+\sinh^2\theta=\cosh^2\theta,
$$

we obtain

$$
\sqrt{1+u^2}=\cosh\theta.
$$

Therefore,

$$
\int u^2\sqrt{1+u^2}\,du
=
\int\sinh^2\theta\cosh^2\theta\,d\theta.
$$

Using

$$
\sinh(2\theta)=2\sinh\theta\cosh\theta,
$$

gives

$$
\sinh^2\theta\cosh^2\theta
=
\frac14\sinh^2(2\theta).
$$

Using the identity

$$
\cosh(4\theta)=1+2\sinh^2(2\theta),
$$

gives

$$
\sinh^2(2\theta)
=
\frac{\cosh(4\theta)-1}{2}.
$$

Hence

$$
\int u^2\sqrt{1+u^2}\,du
=
\frac18
\int
\left[\cosh(4\theta)-1\right]d\theta,
$$

so

$$
=
\frac18
\left[
\frac14\sinh(4\theta)-\theta
\right].
$$

Now return to $u$. Since

$$
u=\sinh\theta,
$$

then

$$
\theta=\sinh^{-1}(u).
$$

Also,

$$
\sinh(4\theta)
=
4\sinh\theta\cosh\theta\cosh(2\theta).
$$

Using

$$
\sinh\theta=u,
$$

$$
\cosh\theta=\sqrt{1+u^2},
$$

and

$$
\cosh(2\theta)=1+2u^2,
$$

gives

$$
\frac14\sinh(4\theta)
=
u\sqrt{1+u^2}(1+2u^2)
=
(2u^3+u)\sqrt{1+u^2}.
$$

Therefore,

$$
\int u^2\sqrt{1+u^2}\,du
=
\frac18
\left[
(2u^3+u)\sqrt{1+u^2}
-
\sinh^{-1}(u)
\right].
$$

Applying the limits $0$ to $x$ gives

$$
{
\epsilon_{\rm elec}(k_F)
=
\frac{\epsilon_0}{8}
\left[
(2x^3+x)\sqrt{1+x^2}
-
\sinh^{-1}(x)
\right]
}
$$

where

$$
x=\frac{k_F}{m_ec}.
$$

The lower limit contributes zero.

### Electron degeneracy pressure

For a system with an isotropic distribution of momenta, Sagert gives the pressure as

$$
p=
\frac{1}{3}
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_F}kv\,k^2\,dk.
\tag{21}
$$

The integral runs over all occupied electron momentum states from $k=0$ to the Fermi momentum $k_F$.

The factor $1/3$ arises from isotropy. The electron momenta have no preferred spatial direction, so the momentum contribution is shared equally between the three spatial directions.

For a relativistic electron,

$$
v=\frac{kc^2}{E},
$$

where

$$
E(k)=\sqrt{k^2c^2+m_e^2c^4}.
$$

Therefore,

$$
kv
=
k\left(\frac{kc^2}{E}\right)
=
\frac{k^2c^2}{E}.
$$

Substituting this into the pressure integral gives

$$
p(k_F)
=
\frac{1}{3}
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_F}
\frac{k^4c^2}{E(k)}\,dk,
$$

and hence

$$
p(k_F)
=
\frac{8\pi}{3(2\pi\hbar)^3}
\int_0^{k_F}
(k^2c^2+m_e^2c^4)^{-1/2}c^2k^4\,dk.
$$

#### Dimensionless form

As for the electron energy density, define

$$
u=\frac{k}{m_ec},
\qquad
k=m_ecu,
\qquad
dk=m_ec\,du.
$$

The upper integration limit becomes

$$
x=\frac{k_F}{m_ec}.
$$

Under this substitution,

$$
k^4c^2=m_e^4c^6u^4,
$$

while

$$
\sqrt{k^2c^2+m_e^2c^4}
=
m_ec^2\sqrt{1+u^2}.
$$

Therefore,

$$
\frac{k^4c^2}
{\sqrt{k^2c^2+m_e^2c^4}}
\,dk
=
m_e^4c^5
\frac{u^4}{\sqrt{1+u^2}}
\,du.
$$

Using

$$
\epsilon_0=
\frac{m_e^4c^5}{\pi^2\hbar^3},
$$

the pressure becomes

$$
p(k_F)
=
\frac{\epsilon_0}{3}
\int_0^x
(1+u^2)^{-1/2}u^4\,du.
$$

#### Evaluating the pressure integral

To evaluate

$$
\int\frac{u^4}{\sqrt{1+u^2}}\,du,
$$

use the hyperbolic substitution

$$
u=\sinh\theta.
$$

Then

$$
du=\cosh\theta\,d\theta
$$

and, using

$$
1+\sinh^2\theta=\cosh^2\theta,
$$

we obtain

$$
\sqrt{1+u^2}=\cosh\theta.
$$

Therefore,

$$
\int
\frac{u^4}{\sqrt{1+u^2}}\,du
=
\int\sinh^4\theta\,d\theta.
$$

Using

$$
\sinh^2\theta
=
\frac{\cosh(2\theta)-1}{2},
$$

gives

$$
\sinh^4\theta
=
\frac{1}{4}
\left[
\frac{1}{2}\cosh(4\theta)
-2\cosh(2\theta)
+\frac{3}{2}
\right].
$$

Hence,

$$
\int\sinh^4\theta\,d\theta
=
\frac{1}{32}\sinh(4\theta)
-\frac{1}{4}\sinh(2\theta)
+\frac{3}{8}\theta.
$$

Since

$$
\theta=\sinh^{-1}u,
$$

$$
\sinh(2\theta)=2u\sqrt{1+u^2},
$$

and

$$
\sinh(4\theta)
=
4u\sqrt{1+u^2}(1+2u^2),
$$

the integral becomes

$$
\int
\frac{u^4}{\sqrt{1+u^2}}\,du
=
\frac{1}{8}
\left[
(2u^3-3u)\sqrt{1+u^2}
+
3\sinh^{-1}u
\right].
$$

The lower limit $u=0$ contributes zero. Evaluating at

$$
x=\frac{k_F}{m_ec}
$$

therefore gives

$$
\boxed{
p(k_F)
=
\frac{\epsilon_0}{24}
\left[
(2x^3-3x)\sqrt{1+x^2}
+
3\sinh^{-1}x
\right]
}
\tag{22}
$$

The Fermi momentum $k_F$ is set by the electron number density and therefore by the mass density of the white dwarf.

Increasing the density forces electrons to occupy progressively higher momentum states because of the Pauli exclusion principle. The Fermi momentum therefore increases as the matter is compressed:

$$
\rho\uparrow
\quad\Rightarrow\quad
k_F\uparrow
\quad\Rightarrow\quad
p\uparrow.
$$

The dimensionless quantity

$$
x=\frac{k_F}{m_ec}
$$

also indicates whether the electrons at the Fermi surface are relativistic:

$$
x\ll1
\qquad
\text{nonrelativistic},
$$

$$
x\sim1
\qquad
\text{transition region},
$$

$$
x\gg1
\qquad
\text{relativistic}.
$$

Thus Eq. (22) gives the electron degeneracy pressure across both the nonrelativistic and relativistic regimes.

### Non relativistic electron degeneracy pressure

For non relativistic electrons,

$$
k_F \ll m_e c
$$

so

$$
x = \frac{k_F}{m_e c} \ll 1.
$$

Since $0 \leq u \leq x$, $u$ is also small throughout the integral. Therefore,

$$
(1+u^2)^{-1/2} \approx 1.
$$

Equation (23) can therefore be approximated as

$$
p(k_F)
\approx
\frac{\epsilon_0}{3}
\int_0^{k_F/m_e c} u^4 \, du
$$

which gives

$$
p(k_F)
\approx
\frac{\epsilon_0}{15}
\left(
\frac{k_F}{m_e c}
\right)^5.
$$

Using

$$
\epsilon_0
=
\frac{m_e^4 c^5}{\pi^2 \hbar^3}
$$

and

$$
k_F
=
\hbar
\left(
\frac{3\pi^2 \rho Z}{m_N A}
\right)^{1/3},
$$

gives

$$
p(k_F)
\approx
\frac{\hbar^2}{15\pi^2 m_e}
\left(
\frac{3\pi^2 \rho Z}{m_N A}
\right)^{5/3}.
$$

Therefore, in the non relativistic limit,

$$
\boxed{p \propto \rho^{5/3}}.
$$

The degeneracy pressure therefore increases as the mass density raised to the power $5/3$.

### Non relativistic equation of state

Equation (23) showed that for non relativistic degenerate electrons,

$$
p \propto \rho^{5/3}.
$$

Using the relationship

$$
\rho = \frac{\epsilon}{c^2},
$$

gives

$$
p \propto
\left(
\frac{\epsilon}{c^2}
\right)^{5/3}.
$$

Applying the exponent to both the numerator and denominator,

$$
p \propto
\frac{\epsilon^{5/3}}{c^{10/3}}.
$$

Since $c$ is constant, the factor $1/c^{10/3}$ can be included in the constant of proportionality. Sagert therefore writes

$$
\boxed{
p(\epsilon)
\approx
K_{\rm non\text{-}rel}\epsilon^{5/3}
}
$$

The non relativistic electron degeneracy pressure therefore follows a power law in energy density with exponent $5/3$.

Equation (24) puts the result from Equation (23) into the form $p(\epsilon)$ required by the stellar structure equations.



$$
K_{\rm non\text{-}rel}
=
\frac{\hbar^2}{15\pi^2m_e}
\left(
\frac{3\pi^2Z}{m_Nc^2A}
\right)^{5/3}.
$$

For a fixed composition, $A$ and $Z$ are constant, so $K_{\rm non\text{-}rel}$ is constant.

It collects the constant terms in the non relativistic equation of state, leaving the pressure to vary with energy density as

$$
p(\epsilon) \approx K_{\rm non\text{-}rel}\epsilon^{5/3}.
$$

### Relativistic electron degeneracy pressure

**What does the electron degeneracy pressure become when the electrons are relativistic?**

For relativistic electrons,

$$
k_F \gg m_e c,
$$

so

$$
x = \frac{k_F}{m_e c} \gg 1.
$$

Starting from the pressure integral,

$$
p(k_F)
=
\frac{\epsilon_0}{3}
\int_0^{k_F/m_e c}
(1+u^2)^{-1/2}u^4\,du,
$$

in the relativistic limit,

$$
1+u^2 \approx u^2,
$$

so

$$
(1+u^2)^{-1/2}
\approx
(u^2)^{-1/2}
=
u^{-1}.
$$

The integrand therefore becomes

$$
u^{-1}u^4=u^3.
$$

Thus,

$$
p(k_F)
\approx
\frac{\epsilon_0}{3}
\int_0^{k_F/m_e c}u^3\,du
$$

which gives

$$
p(k_F)
\approx
\frac{\epsilon_0}{12}
\left(
\frac{k_F}{m_e c}
\right)^4.
$$

Using

$$
\epsilon_0
=
\frac{m_e^4c^5}{\pi^2\hbar^3},
$$

gives

$$
p(k_F)
\approx
\frac{c k_F^4}{12\pi^2\hbar^3}.
$$

Using

$$
k_F
=
\hbar
\left(
\frac{3\pi^2\rho Z}{m_NA}
\right)^{1/3},
$$

gives

$$
p
\approx
\frac{\hbar c}{12\pi^2}
\left(
\frac{3\pi^2\rho Z}{m_NA}
\right)^{4/3}.
$$

Therefore, in the relativistic limit,

$$
{p\propto\rho^{4/3}}.
$$

The electron degeneracy pressure therefore increases as the mass density raised to the power $4/3$.

### Relativistic equation of state

**How are pressure and energy density related when the electrons are relativistic?**

Using

$$
\rho=\frac{\epsilon}{c^2},
$$

the relativistic pressure relation

$$
p\propto\rho^{4/3}
$$

becomes

$$
p\propto
\left(
\frac{\epsilon}{c^2}
\right)^{4/3}
=
\frac{\epsilon^{4/3}}{c^{8/3}}.
$$

Sagert therefore writes the relativistic equation of state as

$$
{
p(\epsilon)\approx K_{\rm rel}\epsilon^{4/3}
}
$$

where

$$
{
K_{\rm rel}
=
\frac{\hbar c}{12\pi^2}
\left(
\frac{3\pi^2 Z}{m_Nc^2A}
\right)^{4/3}
}
$$

For a fixed composition, $A$ and $Z$ are constant, so $K_{\rm rel}$ is constant.

Compared with the non relativistic equation of state,

$$
p(\epsilon)\approx K_{\rm non\text{-}rel}\epsilon^{5/3},
$$

the exponent has decreased from $5/3$ to $4/3$. The pressure therefore increases less rapidly with increasing energy density in the relativistic regime.
