# Compact Star Modelling: Glossary and Notation Reference

Notation convention: Sagert often writes central quantities with subscript $0$, whereas our Python code often uses _c. The letter c alone is the speed of light. Unless specified otherwise, calculations use SI units. The electron momentum is denoted $k$ in Sagert, even though many other texts reserve $k$ for wave number.

## Terminology 

| Term | Meaning |
|---|---|
| Compact star | A very dense stellar remnant star. |
| White dwarf | Compact remnant generally supported by electron degeneracy pressure. |
| Neutron star | Extremely compact remnant. |
| Stellar remnant | Material remaining after a star's earlier evolutionary stages. |
| Stellar core | Central region of a star. |
| Core collapse | Rapid gravitational contraction of a stellar core. |
| Core collapse supernova | Explosive collapse of a massive stellar core. |
| Proto neutron star | Hot, newly formed neutron star shortly after collapse. |
| Black hole | Compact region containing an event horizon, beyond which light cannot escape. |
| Nuclear fusion | Joining lighter nuclei into heavier nuclei. |
| Electron capture | Process in which a proton and electron form a neutron and electron neutrino. |
| Neutron drip | Onset of unbound neutrons as increasingly neutron rich nuclei cannot retain all neutrons. |
| Hydrostatic equilibrium | Balance between inward gravity and outward pressure forces. |
| Newtonian gravity | Gravity model based on Newton's law. |
| General relativity (GR) | Theory of gravity based on Einstein's Physics. |
| Spherical symmetry | Physical conditions are independent of direction, depending only on distance from the centre. |
| Isotropy | No preferred direction. |
| Spherical shell | Thin region between radii $r$ and $r+dr$; shell volume is approximately $4\pi r^2dr$. |
| Stellar surface | Outer boundary where pressure drops to zero. |
| Chandrasekhar mass | Limiting mass scale for idealised white dwarfs supported by degenerate electrons, arising from relativistic softening of their EOS; value depends on assumptions and composition. |
| Electron degeneracy pressure | Pressure arising from fermionic state filling. |
| Degenerate matter | Matter whose occupied quantum states are strongly constrained by Pauli exclusion. |
| Fermion | Particle with half integer spin. |
| Pauli exclusion principle | Identical fermions cannot occupy the same complete quantum state. |
| Rest energy | Energy $mc^2$ associated with particle mass even at zero momentum. |
| Mass density | Mass per physical volume, in $\mathrm{kg/m^3}$. |
| Energy density | Energy per physical volume, in $\mathrm{J/m^3}$. |
| Pressure | Force per unit area, measured in pascals. |
| Equation of state (EoS) | Relationship between thermodynamic quantities. |
| Polytropic EOS | Approximation $p=K\epsilon^\gamma$, with constants $K$ and $\gamma$. |
| Nonrelativistic | Speeds much smaller than the speed of light. |
| Relativistic / ultrarelativistic | Speeds much closer to the speed of light. |
| Nonrelativistic polytrope | Electron model with $\gamma=5/3$ and $n=3/2$. |
| Relativistic polytrope | Ultrarelativistic electron model with $\gamma=4/3$ and $n=3$. |
| Ideal gas / ideal Fermi gas | Model neglecting particle interactions beyond quantum statistics where stated. |
| Approximation / limiting regime | Simplified formula valid under stated physical conditions; not generally valid at every density. |
| Contravariant | Components transforming with the coordinate Jacobian |
| Covariant | Components transforming with the inverse coordinate Jacobian |
| Free index | Index identifying the component of an equation |
| Dummy index | Repeated index over which summation is implied |
| Einstein summation convention | Rule that a repeated index, once upstairs and once downstairs, implies a sum |
| Contraction | Summation over an upper and lower tensor index |
| Trace | Contraction of a mixed rank two tensor |
| Diagonal matrix | Matrix whose off diagonal entries are zero |
| Inverse metric | Metric satisfying $g^{\mu\alpha}g_{\alpha\nu}=\delta^\mu{}_\nu$ |
| Coordinate basis | Basis vectors associated with a coordinate system |
| Covariant divergence | Contraction of a covariant derivative with a tensor index |
| Isotropic pressure | Equal physical pressure in every spatial direction |
| Radial conservation | Component of local energy and momentum conservation along the radial coordinate | 
| Pure neutron star | Idealised neutron star consisting entirely of neutrons, supported by neutron degeneracy pressure. |
| Neutron degeneracy pressure | Pressure arising from the occupation of quantum momentum states by neutrons obeying the Pauli exclusion principle. |
| Degenerate neutron gas | Collection of neutrons whose occupation of quantum states is governed primarily by Fermi statistics rather than thermal effects. |
| Zero temperature approximation | Assumption that thermal energy is negligible compared with the relevant degeneracy energy scale. |
| Noninteracting neutron gas | Idealised neutron matter model that neglects interactions between neutrons beyond those imposed by quantum statistics. |
| Particle number density | Number of particles of a particular species per unit physical volume. |
| Fermi momentum | Momentum of the highest occupied particle state at zero temperature. |
| Fermi energy | Energy of the highest occupied particle state at zero temperature. |
| Fermi sphere | Region of momentum space containing all occupied states up to the Fermi momentum at zero temperature. |
| Fermi surface | Boundary separating occupied and unoccupied momentum states at zero temperature. |
| Fermi sea | Collection of occupied fermion states below the Fermi energy at zero temperature. |
| Fermi Dirac distribution | Statistical function giving the mean occupation of a fermionic energy state at a specified temperature and chemical potential. |
| Occupation probability | Probability or mean occupation associated with a particular quantum state. |
| Chemical potential | Thermodynamic quantity describing the change in a system's energy when a particle is added under specified conditions. |
| Spin degeneracy | Number of available spin states associated with a given particle momentum. |
| Momentum space | Mathematical space whose coordinates represent particle momentum components rather than physical positions. |
| Phase space | Mathematical space describing the possible combinations of particle positions and momenta. |
| Phase space cell | Elementary volume associated with counting quantum states in position and momentum space. |
| Quantum state | Complete specification of a particle's allowed quantum properties within the model. |
| Relativistic energy momentum relation | Relationship $E^2=k^2c^2+m^2c^4$, connecting particle energy, momentum and rest mass. |
| Rest mass energy density | Energy per unit volume associated with the rest masses of particles, given by $\rho c^2$. |
| Kinetic energy density | Energy per unit volume associated with particle motion rather than rest mass alone. |
| Central pressure | Pressure at the centre of a stellar equilibrium configuration, written $p_0$ or $p_c$. |
| Central density | Density at the centre of a stellar equilibrium configuration, written $\rho_0$ or $\rho_c$. |
| Central energy density | Energy density at the centre of a stellar model. |
| Enclosed mass | Total gravitational mass contained within a spherical radius $r$, represented by $m(r)$. |
| Gravitational mass | Mass associated with the total energy content of a star and determining its external gravitational field. |
| Stellar structure equations | Coupled differential equations describing the radial variation of enclosed mass and pressure inside a star. |
| Initial conditions | Specified starting values required to solve the stellar structure equations numerically. |
| Boundary conditions | Conditions that a mathematical solution must satisfy at specified locations, including the stellar centre and surface. |
| Regular stellar centre | Central region where physical quantities remain finite and enclosed mass approaches zero as radius approaches zero. |
| Polytropic exponent | Exponent $\gamma$ describing the power law relationship $p=K\epsilon^\gamma$. |
| Polytropic index | Quantity $n=1/(\gamma-1)$ used to characterise polytropic stellar models. |
| Polytropic constant | Coefficient $K$ determining the pressure associated with a given energy density in a polytropic equation of state. |
| Lane Emden equation | Dimensionless differential equation describing the equilibrium density profile of a Newtonian polytropic star. |
| Lane Emden function | Dimensionless function $\theta(\xi)$ describing the density profile through $\rho=\rho_0\theta^n$. |
| Dimensionless radial coordinate | Scaled stellar radius $\xi=r/a$ used in the Lane Emden equation. |
| Polytropic mass coefficient | Dimensionless quantity $\xi_1^2|\theta'(\xi_1)|$ used to determine the total mass of a polytropic star. |
| Mass radius relation | Relationship between the total mass and radius of stellar equilibrium configurations. |
| Equilibrium sequence | Collection of stellar models calculated using the same equation of state while varying a central quantity such as pressure or density. |
| Tolman Oppenheimer Volkoff equation (TOV) | General relativistic equation describing the pressure gradient required for hydrostatic equilibrium in a static, spherically symmetric, isotropic fluid star. |
| Relativistic hydrostatic equilibrium | Balance between pressure gradients and gravity when general relativistic effects are included. |
| Newtonian limit | Regime in which relativistic corrections become negligible and the governing equations approach their Newtonian forms. |
| Schwarzschild radius | Characteristic radius $R_s=2GM/c^2$ associated with a gravitational mass $M$. |
| Stellar compactness | Dimensionless measure $2GM/(Rc^2)$ comparing a star's Schwarzschild radius with its physical radius. |
| Local compactness | Dimensionless quantity $2Gm(r)/(rc^2)$ describing compactness at a radius inside the star. |
| Relativistic fluid inertia correction | TOV factor $1+p/\epsilon$, accounting for the contribution of pressure to relativistic fluid inertia. |
| Pressure contribution to gravity | TOV correction involving $4\pi r^3p/(mc^2)$, reflecting the gravitational influence of pressure. |
| Spacetime curvature correction | TOV factor $(1-2Gm/(rc^2))^{-1}$, incorporating the influence of relativistic geometry on hydrostatic equilibrium. |
| Spacetime | Four dimensional description combining spatial coordinates and time. |
| Spacetime curvature | Geometrical property describing how spacetime differs from flat Minkowski spacetime. |
| Metric tensor | Tensor describing spacetime intervals, lengths and the relationship between coordinate components. |
| Minkowski metric | Metric describing flat spacetime in special relativity. |
| Einstein field equations | Equations relating spacetime curvature to the energy and momentum content of matter. |
| Einstein tensor | Tensor constructed from spacetime curvature and appearing in Einstein's field equations. |
| Stress energy tensor | Tensor describing energy density, momentum density, energy transport and stresses within matter. |
| Perfect fluid | Idealised fluid characterised by energy density and isotropic pressure, with no viscosity or heat flow in its local rest frame. |
| Normal stress | Force per unit area acting perpendicular to a surface. |
| Shear stress | Force per unit area acting parallel to a surface. |
| Anisotropic pressure | Pressure whose radial and tangential components are not necessarily equal. |
| Radial pressure | Pressure acting in the radial direction within a stellar fluid. |
| Tangential pressure | Pressure acting in directions perpendicular to the radial direction. |
| Four velocity | Relativistic vector describing the rate of change of spacetime position with proper time. |
| Proper time | Time measured by a clock moving along a particular worldline. |
| Christoffel symbols | Connection coefficients describing how coordinate basis vectors change and appearing in covariant derivatives. |
| Covariant derivative | Derivative accounting for both changes in tensor components and changes in the coordinate basis. |
| Riemann curvature tensor | Tensor describing the curvature of spacetime. |
| Ricci tensor | Tensor obtained by contracting the Riemann curvature tensor. |
| Ricci scalar | Scalar curvature quantity obtained by contracting the Ricci tensor with the inverse metric. |
| Radial oscillation | Perturbation in which stellar fluid elements move radially about an equilibrium configuration. |
| Stellar perturbation | Small disturbance applied to a stellar equilibrium configuration to investigate its response. |
| Radial oscillation mode | Particular pattern of radial stellar displacement associated with an oscillation frequency. |
| Fundamental radial mode | Lowest frequency radial oscillation mode, labelled by index $0$. |
| Excited radial mode | Radial oscillation mode with a higher mode index than the fundamental mode. |
| Squared mode frequency | Quantity $\omega_n^2$, whose sign determines whether a radial mode is oscillatory or can grow exponentially. |
| Dynamical stability | Ability of an equilibrium configuration to resist small dynamical perturbations without developing growing unstable modes. |
| Radial instability | Instability in which a small radial disturbance contains a mode that grows rather than remaining oscillatory. |
| Stable oscillation | Bounded oscillatory response of a perturbation associated with positive squared mode frequency. |
| Exponential instability | Behaviour in which a perturbation grows exponentially in time, associated with negative squared mode frequency. |
| Turning point | Maximum or minimum along a stellar equilibrium sequence, potentially associated with a change in radial stability. |
| Critical point | Configuration at which a stability mode may pass through zero squared frequency. |
| First variation | First order change in a quantity under a small perturbation, such as $\delta M$. |
| Second variation | Second order change in a quantity under a small perturbation, such as $\delta^2M$. |
| Constrained equilibrium | Equilibrium investigated under perturbations that preserve specified conserved quantities, such as baryon number. |
