# Compact Star Modelling: Glossary and Notation Reference

Notation convention: Sagert often writes central quantities with subscript $0$, whereas our Python code often uses _c. The letter c alone is the speed of light. Unless specified otherwise, calculations use SI units. The electron momentum is denoted $k$ in Sagert, even though many other texts reserve $k$ for wave number.

## Terminology 

| Term | Meaning |
|---|---|
| Compact star | A very dense stellar remnant star. |
| White dwarf | Compact remnant generally supported by electron degeneracy pressure. |
| Neutron star | Extremely compact remnant. |
| Stellar remnant | Material remaining after a star's earlier evolutionary stages. |
| Stellar core | Central region of a star. |
| Core collapse | Rapid gravitational contraction of a stellar core. |
| Core collapse supernova | Explosive collapse of a massive stellar core. |
| Proto neutron star | Hot, newly formed neutron star shortly after collapse. |
| Black hole | Compact region containing an event horizon, beyond which light cannot escape. |
| Nuclear fusion | Joining lighter nuclei into heavier nuclei. |
| Electron capture | Process in which a proton and electron form a neutron and electron neutrino. |
| Neutron drip | Onset of unbound neutrons as increasingly neutron rich nuclei cannot retain all neutrons. |
| Hydrostatic equilibrium | Balance between inward gravity and outward pressure forces. |
| Newtonian gravity | Gravity model based on Newton's law. |
| General relativity (GR) | Theory of gravity based on Einstein's Physics. |
| Spherical symmetry | Physical conditions are independent of direction, depending only on distance from the centre. |
| Isotropy | No preferred direction. |
| Spherical shell | Thin region between radii $r$ and $r+dr$; shell volume is approximately $4\pi r^2dr$. |
| Stellar surface | Outer boundary where pressure drops to zero. |
| Chandrasekhar mass | Limiting mass scale for idealised white dwarfs supported by degenerate electrons, arising from relativistic softening of their EOS; value depends on assumptions and composition. |
| Electron degeneracy pressure | Pressure arising from fermionic state filling. |
| Degenerate matter | Matter whose occupied quantum states are strongly constrained by Pauli exclusion. |
| Fermion | Particle with half integer spin. |
| Pauli exclusion principle | Identical fermions cannot occupy the same complete quantum state. |
| Rest energy | Energy $mc^2$ associated with particle mass even at zero momentum. |
| Mass density | Mass per physical volume, in $\mathrm{kg/m^3}$. |
| Energy density | Energy per physical volume, in $\mathrm{J/m^3}$. |
| Pressure | Force per unit area, measured in pascals. |
| Equation of state (EoS) | Relationship between thermodynamic quantities. |
| Polytropic EOS | Approximation $p=K\epsilon^\gamma$, with constants $K$ and $\gamma$. |
| Nonrelativistic | Speeds much smaller than the speed of light. |
| Relativistic / ultrarelativistic | Speeds much closer to the speed of light. |
| Nonrelativistic polytrope | Electron model with $\gamma=5/3$ and $n=3/2$. |
| Relativistic polytrope | Ultrarelativistic electron model with $\gamma=4/3$ and $n=3$. |
| Ideal gas / ideal Fermi gas | Model neglecting particle interactions beyond quantum statistics where stated. |
| Approximation / limiting regime | Simplified formula valid under stated physical conditions; not generally valid at every density. |
