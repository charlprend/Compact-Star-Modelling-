# Compact Stars: Equation Reference

## Stellar structure

### Equations 1 to 8

**How does the white dwarf hold itself up, and how does its structure change with $r$?**

How the enclosed mass builds up as we move outward:

$$
\frac{dm}{dr}=\frac{4\pi r^2\epsilon(r)}{c^2}
$$

How the pressure changes as we move outward

$$
\frac{dp}{dr}=-\frac{G\epsilon(r)m(r)}{c^2r^2}
$$

## Electron degeneracy and the equation of state

### Big picture question

**Where do the pressure and energy density actually come from?**

### Equation 9

**How likely is a state of energy $E$ to be occupied?**

$$
f(E)=\frac{1}{\exp[(E-\mu)/k_BT]+1}
$$

### Equation 10

**How many occupied particle states are there per unit volume, per unit volume of momentum space?**

$$
\frac{dn_i}{d^3k}=\frac{g}{(2\pi\hbar)^3}f
$$

### Equation 11

**How do we add up all the occupied states across momentum space to obtain the particle number density?**

$$
n_i=\int dn_i=\int\frac{g}{(2\pi\hbar)^3}f\,d^3k
$$

### Equation 12

**At $T=0$, which states are actually occupied?**

$$
f(E)=
\begin{cases}
1 & E\leq E_F\\
0 & E>E_F
\end{cases}
$$

### Equation 13

**What electron number density do we get when we add up all those occupied states?**

$$
n_e=\frac{k_F^3}{3\pi^2\hbar^3}
$$

### Equation 14

**How do we turn electron number density into the mass density of white dwarf matter?**

$$
\rho=nm_N\frac{A}{Z}
$$

### Equation 15

**If we know the mass density, what is the Fermi momentum of the electrons?**

$$
k_F=\hbar\left(\frac{3\pi^2\rho Z}{m_NA}\right)^{1/3}
$$

### Equation 16

**How do the nuclei and electrons contribute to the total energy density of white dwarf matter?**

$$
\epsilon=nm_N\frac{A}{Z}c^2+\epsilon_{\mathrm{elec}}(k_F)
$$

### Equation 17

**For an electron occupying a momentum state $k$, how much energy does it have?**

$$
E(k)=\sqrt{k^2c^2+m_e^2c^4}
$$

### Equation 18

**What is the total energy density of all the electrons in the occupied momentum states?**

$$
\epsilon_{\mathrm{elec}}(k_F)
=
\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_F}E(k)k^2\,dk
$$

$$
=
\frac{\epsilon_0}{8}
\left[
(2x^3+x)\sqrt{1+x^2}
-\sinh^{-1}(x)
\right]
$$

### Equations 19 and 20

**What constants and dimensionless variables simplify the electron energy and pressure expressions?**

$$
\epsilon_0=\frac{m_e^4c^5}{\pi^2\hbar^3}
$$

$$
x=\frac{k_F}{m_ec}
$$

### Equation 21

**What is the total pressure of all the electrons in the occupied momentum states?**

$$
p=\frac13\frac{8\pi}{(2\pi\hbar)^3}
\int_0^{k_F}kvk^2\,dk
$$

### Equation 22

**What is the electron degeneracy pressure for a given Fermi momentum?**

$$
p(k_F)=\frac{\epsilon_0}{24}
\left[
(2x^3-3x)\sqrt{1+x^2}
+3\sinh^{-1}(x)
\right]
$$

## Limiting equations of state

### Equation 23

**What does the electron degeneracy pressure become when the electrons are non relativistic?**

$$
p(k_F)
=
\frac{\epsilon_0}{3}
\int_0^{k_F/(m_ec)}
(u^2+1)^{-1/2}u^4\,du
$$

$$
\approx
\frac{\epsilon_0}{15}
\left(\frac{k_F}{m_ec}\right)^5
=
\frac{\hbar^2}{15\pi^2m_e}
\left(\frac{3\pi^2\rho Z}{m_NA}\right)^{5/3}
$$

### Equation 24

**How are pressure and energy density related when the electrons are non relativistic?**

$$
p(\epsilon)\approx K_{\mathrm{nonrel}}\epsilon^{5/3}
$$

### Equation 25

**What is the constant that determines the non relativistic equation of state?**

$$
K_{\mathrm{nonrel}}
=
\frac{\hbar^2}{15\pi^2m_e}
\left(
\frac{3\pi^2Z}{m_Nc^2A}
\right)^{5/3}
$$

### Equation 26

**How are pressure and energy density related when the electrons are relativistic?**

$$
p(\epsilon)\approx K_{\mathrm{rel}}\epsilon^{4/3}
$$

### Equation 27

**What is the constant that determines the relativistic equation of state?**

$$
K_{\mathrm{rel}}
=
\frac{\hbar c}{12\pi^2}
\left(
\frac{3\pi^2Z}{m_Nc^2A}
\right)^{4/3}
$$

### Equation 28

**What general form describes a polytropic equation of state?**

$$
p=K\epsilon^\gamma
$$

## Ultra relativistic electron Fermi gas

### Equations 29 to 34

**How are pressure and energy density related for an ultra relativistic electron Fermi gas whose energy density comes entirely from the electrons?**

$$
p=\frac13\epsilon
$$

## Chandrasekhar mass

### Equation 35

**What equation describes stellar equilibrium after eliminating the enclosed mass function?**

$$
\frac{1}{r^2}\frac{d}{dr}
\left(
\frac{r^2}{\rho}\frac{dp}{dr}
\right)
=-4\pi G\rho
$$

### Equation 36

**How is the polytropic equation of state expressed in terms of mass density?**

$$
p=K\epsilon^\gamma=K\rho^\gamma c^{2\gamma}
$$

### Equation 37

**What is the density boundary condition at the centre of the star?**

$$
\rho(r=0)=\rho_0\neq0
$$

### Equation 38

**What is the density gradient at the centre of the star?**

$$
\left.\frac{d\rho}{dr}\right|_{r=0}=0
$$

### Equations 39 and 40

**What characteristic scale and dimensionless variables simplify the stellar structure equations?**

$$
a=
\left(
\frac{(n+1)K\rho_0^{(1-n)/n}}{4\pi G}
\right)^{1/2}
c^{(n+1)/n}
$$





$$
r=a\xi
\qquad
\rho(r)=\rho_0\theta^n
\qquad
\theta=\theta(r)
$$

### Equation 41

**What equation describes the dimensionless density profile of a polytropic star?**

$$
\frac{1}{\xi^2}\frac{d}{d\xi}
\left(\xi^2\frac{d\theta}{d\xi}\right)
=-\theta^n
$$

### Equation 42

**What is the dimensionless density at the centre of the star?**

$$
\theta(0)=1
$$

### Equation 43

**What is the density profile's gradient at the centre of the star?**

$$
\left.\frac{d\theta}{dr}\right|_{r=0}=0
$$

### Equation 44

**Where does the stellar surface occur, and how do we calculate its physical radius?**

$$
\theta(\xi_1)=0
$$

$$
R=a\xi_1
$$

### Equation 45

**How do we calculate the total stellar mass from the Lane Emden density profile?**

$$
M=4\pi a^3\rho_0
\int_0^{\xi_1}\xi^2\theta^n\,d\xi
$$

$$
M=4\pi a^3\rho_0\xi_1^2
|\theta'(\xi_1)|
$$

$$
M=
4\pi c^{(3n+3)/n}
\left(\frac{(n+1)K}{4\pi G}\right)^{3/2}
\rho_0^{(3-n)/(2n)}
\xi_1^2|\theta'(\xi_1)|
$$

### Equation 46

**How are the mass and radius of a polytropic star related once the central density has been eliminated?**

$$
M=
4\pi c^{(2n+2)/(n-1)}
\left(
\frac{(n+1)K}{4\pi G}
\right)^{n/(n-1)}
\xi_1^{(n-3)/(1-n)}
\xi_1^2|\theta'(\xi_1)|
R^{(3-n)/(1-n)}
$$

### Equation 47

**What polytropic index and Lane Emden constants describe a white dwarf supported by relativistic degenerate electrons?**

$$
\gamma=\frac43
$$

$$
n=3
$$

$$
\xi_1=6.89685
$$

$$
\xi_1^2|\theta'(\xi_1)|=2.01824
$$



### Equation 48

**How does the radius of a relativistic white dwarf depend on its central density?**

$$
R=
\frac12(3\pi)^{1/2}(6.89685)
\left(
\frac{\hbar^{3/2}}
{c^{1/2}G^{1/2}m_em_N\eta}
\right)
\left(
\frac{\rho_{\mathrm{crit}}}{\rho_0}
\right)^{1/3}
$$

### Equation 49

**At what mass density does the electron Fermi momentum become comparable to the electron rest mass momentum?**

$$
\rho_{\mathrm{crit}}
=
\frac{m_N\eta m_e^3c^3}
{3\pi^2\hbar^3}
$$


### Equation 50

**What is the limiting mass of a white dwarf supported by relativistic electron degeneracy pressure?**

$$
M=
\frac12(3\pi)^{1/2}(2.01824)
\left(
\frac{\hbar c}{G}
\right)^{3/2}
\left(
\frac{1}{m_N\eta}
\right)^2
$$

### Equation 51

**What is the Chandrasekhar mass expressed in solar masses?**

$$
M=1.4312
\left(
\frac2\eta
\right)^2M_\odot
$$

For carbon 12,

$$
\eta=\frac{12}{6}=2
$$

and for oxygen 16,

$$
\eta=\frac{16}{8}=2
$$

Therefore, for either composition,

$$
M=1.4312M_\odot
$$
